````python
import datetime
import math
from typing import Dict, List, Any, Tuple
import swisseph as swe

# =====================================================================
# CONSTANTS & VEDIC ASTROLOGY DATASETS
# =====================================================================

RASHIS = [
    "Mesham", "Rishabham", "Mithunam", "Karkataka",
    "Simham", "Kanya", "Thula", "Vrischikam",
    "Dhanus", "Makaram", "Kumbham", "Meenam"
]

RASHI_LORDS = {
    "Mesham": "MA", "Rishabham": "VE", "Mithunam": "ME", "Karkataka": "MO",
    "Simham": "SU", "Kanya": "ME", "Thula": "VE", "Vrischikam": "MA",
    "Dhanus": "JU", "Makaram": "SA", "Kumbham": "SA", "Meenam": "JU"
}

NAKSHATRAS = [
    ("Ashwini", "KE"), ("Bharani", "VE"), ("Krittika", "SU"),
    ("Rohini", "MO"), ("Mrigasira", "MA"), ("Arudra", "RA"),
    ("Punarvasu", "JU"), ("Pushya", "SA"), ("Ashlesha", "ME"),
    ("Magha", "KE"), ("Purva Phalguni", "VE"), ("Uttara Phalguni", "SU"),
    ("Hasta", "MO"), ("Chitra", "MA"), ("Swati", "RA"),
    ("Vishakha", "JU"), ("Anuradha", "SA"), ("Jyeshtha", "ME"),
    ("Mula", "KE"), ("Purva Ashadha", "VE"), ("Uttara Ashadha", "SU"),
    ("Shravana", "MO"), ("Dhanishta", "MA"), ("Shatabhisha", "RA"),
    ("Purva Bhadrapada", "JU"), ("Uttara Bhadrapada", "SA"), ("Revati", "ME")
]

GRAHA_KEYS = {
    swe.SUN: "SU",
    swe.MOON: "MO",
    swe.MARS: "MA",
    swe.MERCURY: "ME",
    swe.JUPITER: "JU",
    swe.VENUS: "VE",
    swe.SATURN: "SA",
    swe.MEAN_NODE: "RA"  # Ketu is calculated 180° opposite Rahu
}

HOUSE_DOMAIN_MAP = {
    "career": [10, 6, 1, 2, 11],
    "marriage": [7, 2, 8, 11],
    "love": [5, 7, 11],
    "wealth": [2, 11, 9, 5, 1],
    "education": [4, 5, 9, 2],
    "health": [1, 6, 8, 12]
}

# =====================================================================
# ASTROLOGICAL CALCULATION ENGINE (Swiss Ephemeris)
# =====================================================================

class VedicEphemerisEngine:
    def __init__(self):
        # Set Lahiri Ayanamsa (Nirayana Sidereal)
        swe.set_sid_mode(swe.SIDM_LAHIRI, 0, 0)

    def calculate_chart(self, year: int, month: int, day: int, hour: float, lat: float, lon: float) -> Dict[str, Any]:
        """Calculates Sidereal (Nirayana) Rasi (D1) and Navamsa (D9) positions."""
        # Universal Time (UT) Julian Day
        julian_day = swe.julday(year, month, day, hour)
        ayanamsa = swe.get_ayanamsa_ut(julian_day)

        # 1. Calculate Ascendant / Lagna
        # Placidus house system used for ascendant degree determination
        houses, ascmc = swe.houses_ex(julian_day, lat, lon, b'P', swe.FLG_SIDEREAL)
        lagna_deg = ascmc[0]
        lagna_rashi_idx = int(lagna_deg // 30)

        chart_placements = {
            "LA": self._process_degree(lagna_deg, "Ascendant / Lagna")
        }

        # 2. Calculate Planets (Rasi D1)
        for planet_id, code in GRAHA_KEYS.items():
            res, _ = swe.calc_ut(julian_day, planet_id, swe.FLG_SIDEREAL)
            lon_deg = res[0]
            chart_placements[code] = self._process_degree(lon_deg, code)

        # 3. Calculate Ketu (180° opposite Rahu)
        rahu_lon = chart_placements["RA"]["degree"]
        ketu_lon = (rahu_lon + 180.0) % 360.0
        chart_placements["KE"] = self._process_degree(ketu_lon, "KE")

        # 4. Calculate Bhavas (Houses) relative to Lagna
        bhavas = {}
        for code, details in chart_placements.items():
            bhava_num = ((details["rashi_index"] - lagna_rashi_idx) % 12) + 1
            details["bhava"] = bhava_num
            bhavas[bhava_num] = bhavas.get(bhava_num, []) + [code]

        # 5. Build Navamsa (D9) Chart
        d9_placements = {}
        for code, details in chart_placements.items():
            d9_placements[code] = details["navamsha_rashi"]

        return {
            "julian_day": julian_day,
            "ayanamsa": ayanamsa,
            "lagna_rashi": RASHIS[lagna_rashi_idx],
            "d1_placements": chart_placements,
            "d9_placements": d9_placements,
            "bhavas": bhavas
        }

    def _process_degree(self, sidereal_deg: float, name: str) -> Dict[str, Any]:
        """Calculates Rashi, Nakshatra, Pada, and Navamsha (D9) from sidereal degree."""
        sidereal_deg = sidereal_deg % 360.0
        rashi_idx = int(sidereal_deg // 30)

        # Nakshatra calculation (27 Nakshatras = 13°20' or 13.3333° each)
        nak_float = sidereal_deg / (13 + 1/3)
        nak_idx = int(nak_float)

        # Pada calculation (4 Padas per Nakshatra = 3°20' or 3.3333° each)
        pada_float = (sidereal_deg % (13 + 1/3)) / (3 + 1/3)
        pada = int(pada_float) + 1

        # Navamsha (D9) Calculation (Total 108 Padas mapped to 12 Rashis sequentially)
        overall_pada_index = int(sidereal_deg // (3 + 1/3))
        navamsha_rashi_idx = overall_pada_index % 12

        nak_name, nak_lord = NAKSHATRAS[nak_idx]

        return {
            "degree": sidereal_deg,
            "rashi_index": rashi_idx,
            "rashi": RASHIS[rashi_idx],
            "nakshatra": nak_name,
            "nakshatra_lord": nak_lord,
            "pada": pada,
            "navamsha_rashi": RASHIS[navamsha_rashi_idx]
        }

# =====================================================================
# SOUTH INDIAN CHART RENDERER (ASCII Grid Format)
# =====================================================================

class SouthIndianChartFormatter:
    """Formats Rasi (D1) and Navamsa (D9) charts into traditional South Indian 4x4 Grid layout."""

    # Fixed positions of Rashis in South Indian Chart
    # Row 0: Meenam (0,0), Mesham (0,1), Rishabham (0,2), Mithunam (0,3)
    # Row 1: Kumbham (1,0), [CENTER], [CENTER], Karkataka (1,3)
    # Row 2: Makaram (2,0), [CENTER], [CENTER], Simham (2,3)
    # Row 3: Dhanus (3,0), Vrischikam (3,1), Thula (3,2), Kanya (3,3)
    GRID_MAP = {
        "Meenam": (0, 0), "Mesham": (0, 1), "Rishabham": (0, 2), "Mithunam": (0, 3),
        "Karkataka": (1, 3), "Simham": (2, 3), "Kanya": (3, 3), "Thula": (3, 2),
        "Vrischikam": (3, 1), "Dhanus": (3, 0), "Makaram": (2, 0), "Kumbham": (1, 0)
    }

    @classmethod
    def render_rasi_chart(cls, chart_data: Dict[str, Any]) -> str:
        grid = [["" for _ in range(4)] for _ in range(4)]

        # Populate Lagna and Planets
        placements = chart_data["d1_placements"]
        for graha_code, details in placements.items():
            rashi = details["rashi"]
            r, c = cls.GRID_MAP[rashi]
            prefix = "[LA]" if graha_code == "LA" else graha_code
            grid[r][c] += f"{prefix} "

        return cls._draw_grid(grid, "SOUTH INDIAN RASI (D1) CHART")

    @classmethod
    def render_navamsa_chart(cls, d9_placements: Dict[str, str]) -> str:
        grid = [["" for _ in range(4)] for _ in range(4)]

        for graha_code, rashi in d9_placements.items():
            r, c = cls.GRID_MAP[rashi]
            prefix = "[LA]" if graha_code == "LA" else graha_code
            grid[r][c] += f"{prefix} "

        return cls._draw_grid(grid, "SOUTH INDIAN NAVAMSA (D9) CHART")

    @classmethod
    def _draw_grid(cls, grid: List[List[str]], title: str) -> str:
        lines = [f"\n+=================== {title} ===================+"]
        for r in range(4):
            line = "|"
            for c in range(4):
                if (r == 1 and c in [1, 2]) or (r == 2 and c in [1, 2]):
                    if r == 1 and c == 1:
                        line += f" {title.center(32)} |"
                    elif r == 1 and c == 2:
                        continue  # Merged center
                    else:
                        continue
                else:
                    val = grid[r][c].strip() if grid[r][c] else " "
                    line += f" {val:<14} |"
            lines.append(line)
            lines.append("+---------------+---------------+---------------+---------------+")
        return "\n".join(lines)

# =====================================================================
# EXPERT SYSTEM PREDICTION ENGINE
# =====================================================================

class VedicExpertSystem:
    def __init__(self, chart_data: Dict[str, Any]):
        self.chart = chart_data
        self.d1 = chart_data["d1_placements"]
        self.d9 = chart_data["d9_placements"]
        self.bhavas = chart_data["bhavas"]
        self.lagna_rashi = chart_data["lagna_rashi"]

    def evaluate_life_domain(self, domain: str) -> Dict[str, Any]:
        """Generates targeted expert predictions based on house lords, planets, and nakshatras."""
        domain = domain.lower().strip()
        target_houses = HOUSE_DOMAIN_MAP.get(domain, [1, 10, 9])

        primary_house = target_houses[0]
        house_lord_code = self._get_house_lord(primary_house)
        lord_placement = self.d1[house_lord_code]

        # House Occupants
        occupants = self.bhavas.get(primary_house, [])

        # Synthesis
        predictions = []
        predictions.append(
            f"Primary {domain.capitalize()} House ({primary_house}th Bhava) falls in {self._get_house_rashi(primary_house)}."
        )
        predictions.append(
            f"House Lord ({house_lord_code}) is placed in Bhava {lord_placement['bhava']} ({lord_placement['rashi']}) "
            f"in {lord_placement['nakshatra']} Nakshatra (Pada {lord_placement['pada']})."
        )

        if occupants:
            predictions.append(f"Planets occupying the {primary_house}th house: {', '.join(occupants)}.")
        else:
            predictions.append(f"The {primary_house}th house has no direct planetary occupants; influenced by lord {house_lord_code}.")

        # Nakshatra-Pada Synthesis from Matrix
        pada_prediction = (
            f"Micro-Dharma in Navamsha ({lord_placement['navamsha_rashi']}): "
            f"Lord {house_lord_code} in {lord_placement['nakshatra']} Pada {lord_placement['pada']} indicates "
            f"focused energy regarding {domain} through {lord_placement['nakshatra_lord']} nakshatra disposition."
        )
        predictions.append(pada_prediction)

        return {
            "domain": domain.capitalize(),
            "primary_house": primary_house,
            "house_lord": house_lord_code,
            "occupants": occupants,
            "summary_prediction": " ".join(predictions)
        }

    def _get_house_rashi(self, bhava_num: int) -> str:
        lagna_idx = RASHIS.index(self.lagna_rashi)
        target_idx = (lagna_idx + bhava_num - 1) % 12
        return RASHIS[target_idx]

    def _get_house_lord(self, bhava_num: int) -> str:
        rashi = self._get_house_rashi(bhava_num)
        return RASHI_LORDS[rashi]

# =====================================================================
# INTERACTIVE BACKEND CLI RUNNER
# =====================================================================

def run_backend_demo():
    print("=================================================================")
    print("      VEDIC ASTROLOGY BACKEND & EXPERT SYSTEM (SOUTH INDIAN)     ")
    print("=================================================================")

    # Input birth details
    print("\nEnter Birth Details:")
    year = int(input("Year (YYYY) [e.g. 1995]: ") or "1995")
    month = int(input("Month (1-12) [e.g. 8]: ") or "8")
    day = int(input("Day (1-31) [e.g. 15]: ") or "15")
    hour = float(input("Hour (24hr decimal, e.g. 14.5 for 14:30): ") or "14.5")
    lat = float(input("Latitude (e.g. 13.0827 for Chennai): ") or "13.0827")
    lon = float(input("Longitude (e.g. 80.2707 for Chennai): ") or "80.2707")

    # Compute Charts
    engine = VedicEphemerisEngine()
    chart_data = engine.calculate_chart(year, month, day, hour, lat, lon)

    # Render South Indian Charts
    rasi_chart_ascii = SouthIndianChartFormatter.render_rasi_chart(chart_data)
    navamsa_chart_ascii = SouthIndianChartFormatter.render_navamsa_chart(chart_data["d9_placements"])

    print(rasi_chart_ascii)
    print(navamsa_chart_ascii)

    # Instantiate Expert System Query Engine
    expert_system = VedicExpertSystem(chart_data)

    print("\n-----------------------------------------------------------------")
    print("                EXPERT SYSTEM LIFE PREDICTION QUERY              ")
    print("-----------------------------------------------------------------")

    while True:
        print("\nAsk a question or select a domain:")
        print("Available domains: career, marriage, love, wealth, education, health (or 'exit')")
        query = input("User Query: ").lower().strip()

        if query == 'exit':
            break

        # Match domain keyword
        matched_domain = None
        for domain in HOUSE_DOMAIN_MAP.keys():
            if domain in query:
                matched_domain = domain
                break

        if not matched_domain:
            matched_domain = "career"  # Default fallback

        result = expert_system.evaluate_life_domain(matched_domain)

        print(f"\n[EXPERT SYSTEM ASSESSMENT: {result['domain'].upper()}]")
        print(result["summary_prediction"])

if __name__ == "__main__":
    run_backend_demo()
''' eof

---

### Core Technical Features

1. **Swiss Ephemeris Calculations:** Uses `pyswisseph` with `SIDM_LAHIRI` sidereal ayanamsa to accurately determine planetary longitudes, house boundaries, ascendant (Lagna), and Rahu/Ketu positions.
2. **South Indian Chart Renderer:** Generates ASCII formatted 4x4 grid charts for both the **Rasi (D1)** and **Navamsa (D9)** charts.
3. **972 Nakshatra-Pada Mapping:** Automatically computes the precise Nakshatra, Pada (1–4), and D9 Navamsha sign position for Lagna and all 9 Grahas.
4. **Expert System Engine:** Automatically maps life question domains (*Career, Marriage, Love, Wealth, Education, Health*) to their corresponding Vedic houses (Bhavas), assesses primary house lords, checks for planetary occupants, and generates targeted predictions.

To run the backend, install `pyswisseph`:
```bash
pip install pyswisseph
python app.py'''
````

````python
%%writefile app.py

import streamlit as st
import datetime
from datetime import timedelta
import pytz
from geopy.geocoders import Nominatim
from timezonefinder import TimezoneFinder
import pandas as pd
import numpy as np
import os
import re
import math

# Try importing swisseph for exact calculations; fallback if not available
try:
    import swisseph as swe
    HAS_SWISSEPH = True
except ImportError:
    HAS_SWISSEPH = False

# ==============================================================================
# 1. PAGE CONFIGURATION & CUSTOM STYLING
# ==============================================================================
st.set_page_config(
    page_title="Vedic Astrology Expert System & Dasa Engine",
    page_icon="🔮",
    layout="wide",
    initial_sidebar_state="expanded"
)

st.markdown("""
<style>
    .main-header {
        font-size: 2.2rem;
        color: #1E3A8A;
        text-align: center;
        font-weight: 700;
        margin-bottom: 0.2rem;
    }
    .sub-header {
        font-size: 1.0rem;
        color: #4B5563;
        text-align: center;
        margin-bottom: 1.5rem;
    }
    .south-chart-container {
        display: grid;
        grid-template-columns: repeat(4, 1fr);
        grid-template-rows: repeat(4, 110px);
        gap: 4px;
        background-color: #1E293B;
        padding: 6px;
        border-radius: 8px;
        max-width: 500px;
        margin: 0 auto;
    }
    .chart-box {
        background-color: #FFFFFF;
        border: 1px solid #CBD5E1;
        border-radius: 4px;
        padding: 4px;
        font-size: 11px;
        display: flex;
        flex-direction: column;
        justify-content: space-between;
        overflow: hidden;
    }
    .chart-box-center {
        grid-column: 2 / span 2;
        grid-row: 2 / span 2;
        background-color: #F8FAFC;
        border: 2px solid #94A3B8;
        display: flex;
        flex-direction: column;
        align-items: center;
        justify-content: center;
        font-weight: bold;
        color: #1E3A8A;
        font-size: 14px;
        text-align: center;
    }
    .rashi-label {
        font-size: 9px;
        font-weight: bold;
        color: #64748B;
        text-align: right;
    }
    .planet-tag {
        color: #1E3A8A;
        font-weight: bold;
        font-size: 11px;
    }
    .lagna-tag {
        color: #DC2626;
        font-weight: bold;
        font-size: 11px;
    }
    .prediction-card {
        background-color: #F0F9FF;
        border-left: 5px solid #0284C7;
        padding: 18px;
        border-radius: 6px;
        margin-bottom: 15px;
    }
</style>
""", unsafe_allow_html=True)

# ==============================================================================
# 2. CITIES DATABASE (EXPANDED SRI LANKAN & GLOBAL CITIES)
# ==============================================================================
POPULAR_CITIES = {
    # Sri Lanka
    "Colombo, Sri Lanka": {"lat": 6.9271, "lon": 79.8612, "tz": "Asia/Colombo"},
    "Kandy, Sri Lanka": {"lat": 7.2906, "lon": 80.6337, "tz": "Asia/Colombo"},
    "Jaffna, Sri Lanka": {"lat": 9.6615, "lon": 80.0255, "tz": "Asia/Colombo"},
    "Nuwara Eliya, Sri Lanka": {"lat": 6.9497, "lon": 80.7891, "tz": "Asia/Colombo"},
    "Galle, Sri Lanka": {"lat": 6.0535, "lon": 80.2210, "tz": "Asia/Colombo"},
    "Kurunegala, Sri Lanka": {"lat": 7.4863, "lon": 80.3623, "tz": "Asia/Colombo"},
    "Batticaloa, Sri Lanka": {"lat": 7.7310, "lon": 81.6747, "tz": "Asia/Colombo"},
    "Trincomalee, Sri Lanka": {"lat": 8.5874, "lon": 81.2152, "tz": "Asia/Colombo"},
    "Ampara, Sri Lanka": {"lat": 7.2912, "lon": 81.6724, "tz": "Asia/Colombo"},
    "Badulla, Sri Lanka": {"lat": 6.9934, "lon": 81.0550, "tz": "Asia/Colombo"},
    "Vavuniya, Sri Lanka": {"lat": 8.7514, "lon": 80.4971, "tz": "Asia/Colombo"},
    "Mannar, Sri Lanka": {"lat": 8.9810, "lon": 79.9044, "tz": "Asia/Colombo"},
    "Hatton, Sri Lanka": {"lat": 6.8833, "lon": 80.5961, "tz": "Asia/Colombo"},
    "Ratnapura, Sri Lanka": {"lat": 6.6828, "lon": 80.4016, "tz": "Asia/Colombo"},
    "Matara, Sri Lanka": {"lat": 5.9549, "lon": 80.5550, "tz": "Asia/Colombo"},
    "Anuradhapura, Sri Lanka": {"lat": 8.3114, "lon": 80.4037, "tz": "Asia/Colombo"},
    "Polonnaruwa, Sri Lanka": {"lat": 7.9403, "lon": 81.0188, "tz": "Asia/Colombo"},
    "Hambantota, Sri Lanka": {"lat": 6.1241, "lon": 81.1185, "tz": "Asia/Colombo"},
    "Puttalam, Sri Lanka": {"lat": 8.0362, "lon": 79.8283, "tz": "Asia/Colombo"},

    # India & Global
    "New Delhi, India": {"lat": 28.6139, "lon": 77.2090, "tz": "Asia/Kolkata"},
    "Chennai, India": {"lat": 13.0827, "lon": 80.2707, "tz": "Asia/Kolkata"},
    "Mumbai, India": {"lat": 19.0760, "lon": 72.8777, "tz": "Asia/Kolkata"},
    "London, United Kingdom": {"lat": 51.5074, "lon": -0.1278, "tz": "Europe/London"},
    "New York, USA": {"lat": 40.7128, "lon": -74.0060, "tz": "America/New_York"},
    "Singapore": {"lat": 1.3521, "lon": 103.8198, "tz": "Asia/Singapore"},
    "Dubai, UAE": {"lat": 25.2048, "lon": 55.2708, "tz": "Asia/Dubai"},
    "Sydney, Australia": {"lat": -33.8688, "lon": 151.2093, "tz": "Australia/Sydney"}
}

def resolve_location(location_query):
    if location_query in POPULAR_CITIES:
        data = POPULAR_CITIES[location_query]
        return data["lat"], data["lon"], data["tz"]
    try:
        geolocator = Nominatim(user_agent="vedic_astrology_app_2026")
        loc = geolocator.geocode(location_query, timeout=5)
        if loc:
            tf = TimezoneFinder()
            tz_str = tf.timezone_at(lat=loc.latitude, lng=loc.longitude)
            return loc.latitude, loc.longitude, tz_str or "UTC"
    except Exception:
        pass
    return 6.9271, 79.8612, "Asia/Colombo"

# ==============================================================================
# 3. KNOWLEDGE BASES (GRAHA DESCRIPTIONS, YUTIS & RULES)
# ==============================================================================
GRAHA_DESCRIPTIONS = {
    "SU": [
        "B1: Strong vitality, leadership instinct, dignified ego, eye sensitivity.",
        "B2: Wealth via authority/father, direct speech, potential family friction.",
        "B3: High courage, victorious over rivals, skilled hands, energetic siblings.",
        "B4: Property interest, high self-respect, emotional pride, maternal distance.",
        "B5: Sharp intellect, strong creative drive, high ambition for offspring.",
        "B6: High resistance, dominant over opponents, legal success, heat disorders.",
        "B7: Strong-willed partner, high standard in diplomacy, ego trade-offs.",
        "B8: Interest in occult/research, sudden transformation, vitality care needed.",
        "B9: Righteous nature, connection to government/law, strong moral compass.",
        "B10: Directional Strength (Dig Bala), fast professional rise, executive authority.",
        "B11: High gains through government/influential contacts, stable prosperity.",
        "B12: Interest in solitary pursuits, foreign expenditure, spiritual detachment."
    ],
    "MO": [
        "B1: Sensitive demeanor, public charm, expressive face, fluctuating moods.",
        "B2: Wealth through public ventures, sweet speech, family emotional attachment.",
        "B3: Literary interest, communicative talent, frequent short travel.",
        "B4: Directional Strength (Dig Bala), peace of mind, maternal blessings, comfort.",
        "B5: Intuitive intelligence, poetic mind, strong attachment to children.",
        "B6: Emotional fluctuations under stress, service nature, care with digestive health.",
        "B7: Public attraction, empathetic partner, emotionally responsive relationships.",
        "B8: Unconscious insights, interest in secrets, sudden emotional shifts.",
        "B9: Pilgrimages, philosophical mind, fortune through grace and travel.",
        "B10: Career with public visibility, caregiving/hospitality talent, variable fame.",
        "B11: Wide social network, fluctuating income streams, helpful female friends.",
        "B12: Active imagination, need for solitude, interest in foreign lands/retreats."
    ],
    "MA": [
        "B1: High physical vigor, impulsive drive, leadership ability, physical marks.",
        "B2: Bold speech, financial assertiveness, direct approach in family affairs.",
        "B3: Immense courage, mechanical prowess, protective of siblings, sports ability.",
        "B4: Property ownership, technical mind, energetic home environment, assertiveness.",
        "B5: Sharp tactical intelligence, competitive spirit, bold investment style.",
        "B6: Shatters opposition (Shatru Hanta), high physical stamina, legal triumph.",
        "B7: Dynamic partner, high passion, requires patience in relationships.",
        "B8: Interest in research/engineering, transformation through intense experiences.",
        "B9: Passionate belief system, energetic defender of truth, independent thinker.",
        "B10: Directional Strength (Dig Bala), top performance in engineering/management.",
        "B11: Large financial gains through effort, competitive edge, influential allies.",
        "B12: High energy spent in private pursuits, interest in foreign defense/travel."
    ],
    "ME": [
        "B1: Directional Strength (Dig Bala), sharp intellect, youthful look, eloquent speech.",
        "B2: Financial literacy, skilled in negotiation, family business aptitude.",
        "B3: Outstanding writing/media talent, fast learning curve, active networking.",
        "B4: Intellectual home environment, interest in education/real estate analysis.",
        "B5: Analytical capacity, interest in astrology/mathematics, sharp memory.",
        "B6: Analytical problem solving, debate champion, careful health tracking.",
        "B7: Intellectual companion, diplomatic skill, youthful business partner.",
        "B8: Deep analytical research, interest in codes/cryptography, secret knowledge.",
        "B9: Higher learning, publishing aptitude, philosophical writing, broad outlook.",
        "B10: Corporate management, administrative talent, multi-tasking career.",
        "B11: Diverse income sources, strategic networking, commercial success.",
        "B12: Research in solitude, expenditure on education, subtle analytical mind."
    ],
    "JU": [
        "B1: Directional Strength (Dig Bala), immense wisdom, moral clarity, sound health.",
        "B2: Wise speech, financial abundance, family prosperity, ethical values.",
        "B3: Philosophical advisor, constructive communication, supportive brothers.",
        "B4: Domestic contentment, large house, maternal blessings, vehicles.",
        "B5: Exceptional intelligence (Prajna), spiritual merit, honored children.",
        "B6: Overcomes obstacles through wisdom, legal victory, service mentality.",
        "B7: Devoted/wise partner, high social reputation, balanced contracts.",
        "B8: Unlocks hidden wisdom, long life, seamless legacy asset inheritance.",
        "B9: Natural house of fortune (Bhagya), higher teacher, spiritual devotion.",
        "B10: Highly respected career, advisory roles, judicial/executive success.",
        "B11: Steady accumulation of wealth, elite mentorship, noble aspirations.",
        "B12: Higher spiritual liberation (Moksha), charitable nature, foreign peace."
    ],
    "VE": [
        "B1: Charming personality, refined taste, artistic presence, graceful health.",
        "B2: Luxurious speech, wealth through arts/media, pleasant family life.",
        "B3: Artistic talent in music/writing, sweet relationship with siblings.",
        "B4: Directional Strength (Dig Bala), luxury vehicles, beautiful home, comfort.",
        "B5: Artistic intelligence, romance, skill in design and entertainment.",
        "B6: Gracious handling of conflict, diplomatic resolution of disputes.",
        "B7: Romantic/refined spouse, harmonious business partnerships, public success.",
        "B8: Legacy wealth, smooth joint finances, interest in esoteric arts.",
        "B9: Appreciation for cultural arts, fortune through partnerships/travel.",
        "B10: Career in luxury/arts/diplomacy, high social prestige, pleasant work environment.",
        "B11: Wealth from artistic/female networks, fulfillment of refined desires.",
        "B12: Exaltation house effect (in Pisces), high comfort, expenditure on art/luxury."
    ],
    "SA": [
        "B1: Serious outlook, disciplined nature, slow & steady success, endurance.",
        "B2: Conservative spending, measured speech, delayed accumulation of wealth.",
        "B3: Great patience, persistent effort, strong hands, strategic thinking.",
        "B4: Traditional home, duty toward mother, property through hard work.",
        "B5: Deep focus, structured study, serious attitude toward responsibilities.",
        "B6: Dominates long-term enemies, exceptional work ethic, stamina in crises.",
        "B7: Directional Strength (Dig Bala), mature/loyal partner, enduring contracts.",
        "B8: Exceptional longevity, interest in history/geology, deep endurance.",
        "B9: Respect for traditional wisdom, methodical path to fortune, discipline.",
        "B10: Natural house of career, high organizational power, durable leadership.",
        "B11: Massive wealth built over time, loyal long-term friends, stable gains.",
        "B12: Solitary discipline, low material desires, interest in spiritual austerity."
    ],
    "RA": [
        "B1: Unconventional magnetic personality, innovative drive, distinct worldview.",
        "B2: Unique wealth streams, unconventional speech, ambition for resources.",
        "B3: Extraordinary courage, mastery over digital/foreign communications.",
        "B4: Non-traditional living arrangements, foreign real estate interests.",
        "B5: Unconventional intellect, original creative ideas, speculative skill.",
        "B6: Unbeatable against opponents, triumphs in complex legal battles.",
        "B7: Unique/foreign partner, breakthrough approach to partnerships.",
        "B8: Deep psychological insight, sudden discoveries, esoteric interests.",
        "B9: Questions traditions, seeks knowledge from foreign/unconventional paths.",
        "B10: Sudden rise to prominent career status, media appeal, disruptive innovation.",
        "B11: Multiple massive income sources, vast global network, ambitious goals.",
        "B12: Foreign residency, interest in dreams, private spiritual exploratory mind."
    ],
    "KE": [
        "B1: Spiritual aura, intuitive nature, detached identity, self-inquiry.",
        "B2: Minimalist speech, detachment from material hoarding, intuitive values.",
        "B3: Subtle non-verbal skills, courage in quiet ways, deep internal focus.",
        "B4: Simple home preference, intuitive connection to roots, emotional detachment.",
        "B5: Deep abstract intelligence, karmic insights, unique spiritual mantras.",
        "B6: Instinctive crisis management, detachment from conflicts.",
        "B7: Spiritual/introverted partner, search for deeper soul connection.",
        "B8: Esoteric mastery, mastery over occult sciences, transformational liberation.",
        "B9: Direct spiritual insight, non-materialistic philosophy, pilgrimage.",
        "B10: Unconventional career path, indifference to titles, specialized skill.",
        "B11: Selective circle of friends, spiritual gains over material pursuits.",
        "B12: Natural house of Moksha for Ketu, ultimate spiritual liberation, deep meditation."
    ]
}

YUTI_DESCRIPTIONS = [
    ("SU", "MO", "Amavasya Yuti", "High internal focus, strong alignment of soul & mind, introverted drive.", ["B1", "B5", "B9", "B10"]),
    ("SU", "MA", "Agni-Tejas Yoga", "Immense energy, physical vitality, executive leadership, strong competitive edge.", ["B3", "B6", "B10"]),
    ("SU", "ME", "Budhaditya Yoga", "High intellect, professional success, administrative skill, sharp speech.", ["B1", "B4", "B7", "B10", "B11"]),
    ("SU", "JU", "Guru-Aditya Yoga", "Moral authority, legal wisdom, favored by elders/government, high ethics.", ["B1", "B5", "B9", "B10"]),
    ("SU", "VE", "Venus Combustion / Yuti", "Artistic drive, passionate nature, high luxury ambition, requires relationship care.", ["B1", "B5", "B10"]),
    ("SU", "SA", "Pitri Dosha / Sun-Saturn", "Ego friction with authority, immense discipline under pressure, hard-earned triumph.", ["B3", "B6", "B10", "B11"]),
    ("SU", "RA", "Grahan Yoga (Solar)", "High ambition, sudden political or public rise, quest for unique recognition.", ["B3", "B6", "B10", "B11"]),
    ("SU", "KE", "Solar Moksha Yuti", "Spiritual detachment from fame, deep introspection, self-inquiry, research capacity.", ["B9", "B12"]),
    ("MO", "MA", "Chandra-Mangala Yoga", "Dynamic financial drive, success in real estate/trade, emotional intensity.", ["B2", "B9", "B10", "B11"]),
    ("MO", "ME", "Chandra-Budha Yoga", "Quick wit, commercial acumen, literary skill, highly adaptable intelligence.", ["B1", "B2", "B5", "B10"]),
    ("MO", "JU", "Gaja-Kesari Yoga", "Lasting reputation, moral integrity, financial stability, high public protection.", ["B1", "B4", "B7", "B9", "B10"]),
    ("MO", "VE", "Chandra-Shukra Yoga", "Refined artistic tastes, personal charm, success in luxury/hospitality, peace.", ["B1", "B4", "B7", "B11"]),
    ("MO", "SA", "Vish Yoga / Punarphoo", "Emotional discipline, mental endurance, inclination toward solitary work.", ["B3", "B6", "B10", "B11"]),
    ("MO", "RA", "Kemu-Rahu Yuti", "High imagination, psychological depth, intense intuition, creative innovation.", ["B3", "B6", "B10", "B11"]),
    ("MO", "KE", "Kemu-Ketu Yuti", "Deep emotional detachment, psychic sensitivity, inclination toward meditation.", ["B8", "B9", "B12"]),
    ("MA", "ME", "Krodha / Engineering Yuti", "Sharp argumentative skill, technical talent, quick reflexes, debate mastery.", ["B3", "B6", "B10"]),
    ("MA", "JU", "Guru-Mangala Yoga", "Righteous action, legal victory, executive courage, protective instinct.", ["B1", "B3", "B5", "B9", "B10"]),
    ("MA", "VE", "Kama Yoga / Passion Yuti", "High creative energy, magnetic attraction, passion for arts/design, charm.", ["B1", "B5", "B7", "B10"]),
    ("MA", "SA", "Agni-Maruta / Collision", "Tremendous stamina, mechanical prowess, persistence through obstacles.", ["B3", "B6", "B10", "B11"]),
    ("MA", "RA", "Angarak Yoga", "High impulse, breakthrough explosive energy, success in high-risk engineering/defense.", ["B3", "B6", "B10", "B11"]),
    ("MA", "KE", "Kulkarna Yuti", "Precision skill, surgical efficiency, deep technical focus, sudden action.", ["B3", "B6", "B11"]),
    ("ME", "JU", "Saraswati Yoga Component", "Scholarly wisdom, financial acumen, teaching ability, literary mastery.", ["B1", "B2", "B5", "B9", "B10"]),
    ("ME", "VE", "Mahalaxmi / Design Yoga", "Elegance in speech, artistic & literary success, diplomatic charm.", ["B1", "B2", "B4", "B5", "B10"]),
    ("ME", "SA", "Nipuna / Structured Mind", "Methodical research, scientific orientation, organized thinking, endurance.", ["B2", "B3", "B6", "B10", "B11"]),
    ("ME", "RA", "Research / Technical Yuti", "Breakthrough technological talent, foreign trade skill, innovative media skill.", ["B3", "B6", "B10", "B11"]),
    ("ME", "KE", "Codes / Micro-Intellect Yuti", "Abstract coding talent, intuitive logic, deep research capability.", ["B5", "B8", "B12"]),
    ("JU", "VE", "Dvi-Guru Yoga", "Combined wisdom of laws & diplomacy, vast wealth potential, balanced guidance.", ["B1", "B4", "B5", "B9", "B11"]),
    ("JU", "SA", "Brahma / Saturn-Jupiter", "Balanced judgment, long-term wealth stability, philosophical maturity.", ["B1", "B5", "B9", "B10", "B11"]),
    ("JU", "RA", "Guru-Chandal Yoga", "Unconventional wisdom, questioning dogma, global philosophical reach.", ["B3", "B6", "B10", "B11"]),
    ("JU", "KE", "Gyan Yoga", "Highest spiritual knowledge, natural intuition, detachment from material vanity.", ["B5", "B9", "B12"]),
    ("VE", "SA", "Kalanidhi Component", "Durable artistic skill, success in fashion/architecture industry, steady finance.", ["B2", "B3", "B7", "B10", "B11"]),
    ("VE", "RA", "Asura / Luxury Expansion", "Unconventional aesthetic, international fame in media/arts, high luxury ambition.", ["B1", "B5", "B7", "B10", "B11"]),
    ("VE", "KE", "Detached Aesthetics", "Subtle artistic style, spiritual love, non-materialistic view of relationships.", ["B9", "B12"]),
    ("SA", "RA", "Shani-Rahu / Karmic Yuti", "Technical mastery, heavy structural discipline, resilience under extreme conditions.", ["B3", "B6", "B10", "B11"]),
    ("SA", "KE", "Shani-Ketu / Austerity", "Strict discipline, minimalist lifestyle, deep research, spiritual endurance.", ["B3", "B9", "B12"])
]

def generate_phalaphala(graha_code, graha_name, rashi, nakshatra, pada, navamsha, nak_lord):
    if graha_code == "SU":
        if rashi == "Mesham" and nakshatra == "Krittika" and pada == 1:
            return "Deep Exalted Peak (10 deg), supreme authority, legal victory, regal honor, high vitality."
        elif rashi == "Thula":
            return "Debilitated placement, requires ego restraint, diplomacy challenges, public service focus."
        elif rashi == "Simham":
            return "Own / Moolatrikona placement, natural executive leadership, high vitality, commanding presence."
        else:
            return f"Leadership and willpower expressed through {rashi} traits, guided by {nak_lord} energy in {navamsha} Navamsha."

    elif graha_code == "MO":
        if rashi == "Rishabham":
            return "Exalted placement, mental peace, emotional stability, wealth accumulation, luxury."
        elif rashi == "Vrischikam":
            return "Debilitated placement, emotional intensity, psychological depth, transformation."
        elif rashi == "Karkataka":
            return "Own sign placement, deep caretaking, high intuition, strong maternal connection."
        else:
            return f"Emotional expression and public mind tuned to {rashi}, shaped by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "MA":
        if rashi == "Makaram":
            return "Exalted placement, high strategic endurance, engineering skill, disciplined courage."
        elif rashi == "Karkataka":
            return "Debilitated placement, emotional reactivity, protective instincts, needs energy canalization."
        elif rashi in ["Mesham", "Vrischikam"]:
            return "Own sign placement, bold drive, pioneering action, high competitive stamina."
        else:
            return f"Dynamic drive and assertiveness channeled through {rashi}, influenced by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "ME":
        if rashi == "Kanya":
            return "Exalted / Moolatrikona placement, sharp analytical mind, technical perfectionism, business acumen."
        elif rashi == "Meenam":
            return "Debilitated placement, abstract intuitive thinking, creative imagination, poetic skill."
        elif rashi == "Mithunam":
            return "Own sign placement, high eloquence, commercial adaptability, media dexterity."
        else:
            return f"Intellectual and commercial skills expressed through {rashi}, under {nak_lord} energy in {navamsha} Navamsha."

    elif graha_code == "JU":
        if rashi == "Karkataka":
            return "Exalted placement, benevolent wisdom, high moral standing, expansive fortune, protective advisor."
        elif rashi == "Makaram":
            return "Debilitated placement, pragmatic philosophy, material focus, conservative ethics."
        elif rashi in ["Dhanus", "Meenam"]:
            return "Own sign placement, high spiritual learning, advisory authority, philosophical depth."
        else:
            return f"Wisdom, growth, and moral guidance manifested through {rashi}, guided by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "VE":
        if rashi == "Meenam":
            return "Exalted placement, unconditional devotion, high artistic elegance, refined luxury."
        elif rashi == "Kanya":
            return "Debilitated placement, meticulous aesthetic taste, analytical approach to relationships."
        elif rashi in ["Rishabham", "Thula"]:
            return "Own sign placement, personal charm, artistic talent, financial diplomacy, harmony."
        else:
            return f"Creative, romantic, and diplomatic qualities refined in {rashi}, shaped by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "SA":
        if rashi == "Thula":
            return "Exalted placement, highest justice, diplomatic endurance, long-term organizational success."
        elif rashi == "Mesham":
            return "Debilitated placement, lessons in patience, impulsive resistance, hard-earned growth."
        elif rashi in ["Makaram", "Kumbham"]:
            return "Own / Moolatrikona placement, deep discipline, mass leadership, technical stamina."
        else:
            return f"Structured discipline, duty, and endurance executed through {rashi}, influenced by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "RA":
        return f"Amplified material drive and innovative ambition in {rashi}, seeking expansion in {nakshatra} ({navamsha} Navamsha)."

    elif graha_code == "KE":
        return f"Spiritual detachment, intuitive research, and karmic insights in {rashi}, rooted in {nakshatra} ({navamsha} Navamsha)."

# ==============================================================================
# 4. VEDIC ASTROLOGY CALCULATIONS & DASA TIMING ENGINE
# ==============================================================================
RASHIS = [
    "Mesham", "Rishabham", "Mithunam", "Karkataka",
    "Simham", "Kanya", "Thula", "Vrischikam",
    "Dhanus", "Makaram", "Kumbham", "Meenam"
]

NAKSHATRAS_LIST = [
    ("Ashwini", "Ketu"), ("Bharani", "Venus"), ("Krittika", "Sun"),
    ("Rohini", "Moon"), ("Mrigashira", "Mars"), ("Ardra", "Rahu"),
    ("Punarvasu", "Jupiter"), ("Pushya", "Saturn"), ("Ashlesha", "Mercury"),
    ("Magha", "Ketu"), ("Purva Phalguni", "Venus"), ("Uttara Phalguni", "Sun"),
    ("Hasta", "Moon"), ("Chitra", "Mars"), ("Swati", "Rahu"),
    ("Vishakha", "Jupiter"), ("Anuradha", "Saturn"), ("Jyeshta", "Mercury"),
    ("Mula", "Ketu"), ("Purvashadha", "Venus"), ("Uttarashadha", "Sun"),
    ("Shravana", "Moon"), ("Dhanishta", "Mars"), ("Shatabhishak", "Rahu"),
    ("Purva Bhadra", "Jupiter"), ("Uttara Bhadra", "Saturn"), ("Revati", "Mercury")
]

DASA_ORDER = [
    ("Ketu", 7), ("Venus", 20), ("Sun", 6), ("Moon", 10),
    ("Mars", 7), ("Rahu", 18), ("Jupiter", 16), ("Saturn", 19), ("Mercury", 17)
]
DASA_YEARS = dict(DASA_ORDER)
DASA_LORDS = [lord for lord, _ in DASA_ORDER]

GRAHA_KEYS = {
    "SU": "Sun", "MO": "Moon", "MA": "Mars", "ME": "Mercury",
    "JU": "Jupiter", "VE": "Venus", "SA": "Saturn", "RA": "Rahu", "KE": "Ketu"
}

def calculate_julian_day(dt):
    year, month, day = dt.year, dt.month, dt.day
    hour = dt.hour + dt.minute / 60.0 + dt.second / 3600.0
    if month <= 2:
        year -= 1
        month += 12
    A = math.floor(year / 100)
    B = 2 - A + math.floor(A / 4)
    jd = math.floor(365.25 * (year + 4716)) + math.floor(30.6001 * (month + 1)) + day + B - 1524.5 + (hour / 24.0)
    return jd

def compute_positions(birth_dt_utc, lat, lon):
    jd = calculate_julian_day(birth_dt_utc)

    if HAS_SWISSEPH:
        swe.set_sid_mode(swe.SIDM_LAHIRI)
        cusps, ascmc = swe.houses_ex(jd, lat, lon, b'P', swe.FLG_SIDEREAL)
        lagna_deg = ascmc[0] % 360.0

        planet_map = {
            "SU": swe.SUN, "MO": swe.MOON, "MA": swe.MARS, "ME": swe.MERCURY,
            "JU": swe.JUPITER, "VE": swe.VENUS, "SA": swe.SATURN, "RA": swe.MEAN_NODE
        }
        positions = {}
        for p_code, p_id in planet_map.items():
            res, flag = swe.calc_ut(jd, p_id, swe.FLG_SIDEREAL)
            positions[p_code] = res[0] % 360.0
        positions["KE"] = (positions["RA"] + 180.0) % 360.0
    else:
        # Fallback Lahiri calculation
        T = (jd - 2451545.0) / 36525.0
        ayanamsa = 23.85 + 1.396 * T
        days_since_j2000 = jd - 2451545.0

        positions = {
            "SU": (280.46 + 0.9856474 * days_since_j2000 - ayanamsa) % 360.0,
            "MO": (218.32 + 13.176396 * days_since_j2000 - ayanamsa) % 360.0,
            "MA": (355.43 + 0.5240330 * days_since_j2000 - ayanamsa) % 360.0,
            "ME": (252.25 + 1.6021300 * days_since_j2000 - ayanamsa) % 360.0,
            "JU": (34.35 + 0.0830910 * days_since_j2000 - ayanamsa) % 360.0,
            "VE": (181.98 + 1.6021300 * days_since_j2000 - ayanamsa) % 360.0,
            "SA": (50.08 + 0.0334596 * days_since_j2000 - ayanamsa) % 360.0,
            "RA": (125.04 - 0.0529539 * days_since_j2000 - ayanamsa) % 360.0
        }
        positions["KE"] = (positions["RA"] + 180.0) % 360.0
        lst = (100.46 + 0.985647366 * days_since_j2000 + lon + (birth_dt_utc.hour + birth_dt_utc.minute/60.0)*15) % 360.0
        lagna_deg = (lst - ayanamsa) % 360.0

    chart_details = {}
    lagna_rashi_idx = int(lagna_deg // 30)
    chart_details["LA"] = {
        "deg": lagna_deg,
        "rashi_idx": lagna_rashi_idx,
        "rashi": RASHIS[lagna_rashi_idx],
        "navamsha_rashi_idx": int((lagna_deg * 9 / 30) % 12),
        "navamsha_rashi": RASHIS[int((lagna_deg * 9 / 30) % 12)]
    }

    for p_code, deg in positions.items():
        rashi_idx = int(deg // 30)
        overall_pada_idx = int(deg // (30 / 9))
        nak_idx = overall_pada_idx // 4
        pada = (overall_pada_idx % 4) + 1
        navamsha_rashi_idx = overall_pada_idx % 12
        house_num = ((rashi_idx - lagna_rashi_idx) % 12) + 1

        chart_details[p_code] = {
            "deg": deg,
            "rashi_idx": rashi_idx,
            "rashi": RASHIS[rashi_idx],
            "nakshatra": NAKSHATRAS_LIST[nak_idx % 27][0],
            "nakshatra_lord": NAKSHATRAS_LIST[nak_idx % 27][1],
            "pada": pada,
            "navamsha_rashi_idx": navamsha_rashi_idx,
            "navamsha_rashi": RASHIS[navamsha_rashi_idx],
            "house": house_num,
            "overall_pada_idx": overall_pada_idx
        }

    return chart_details

def calculate_full_dasa_tree(dob, moon_degree):
    arc_per_nakshatra = 13.333333333333334
    nakshatra_idx = int(moon_degree // arc_per_nakshatra) % 27
    degree_in_nakshatra = moon_degree % arc_per_nakshatra
    traversed_fraction = degree_in_nakshatra / arc_per_nakshatra

    nakshatra_name = NAKSHATRAS_LIST[nakshatra_idx][0]
    birth_lord = DASA_LORDS[nakshatra_idx % 9]
    remaining_first_dasa_years = DASA_YEARS[birth_lord] * (1.0 - traversed_fraction)

    start_index = DASA_LORDS.index(birth_lord)
    current_date = dob
    dasa_tree = []

    for i in range(9):
        m_lord = DASA_LORDS[(start_index + i) % 9]
        m_duration_years = remaining_first_dasa_years if i == 0 else DASA_YEARS[m_lord]
        m_start = current_date
        m_end = m_start + timedelta(days=m_duration_years * 365.25)
        current_date = m_end

        bhuktis = []
        b_start_idx = DASA_LORDS.index(m_lord)

        for j in range(9):
            b_lord = DASA_LORDS[(b_start_idx + j) % 9]
            b_duration_years = (DASA_YEARS[m_lord] * DASA_YEARS[b_lord]) / 120.0
            if i == 0:
                b_duration_years *= (remaining_first_dasa_years / DASA_YEARS[m_lord])

            b_start = m_start if j == 0 else bhuktis[-1]["end_date"]
            b_end = b_start + timedelta(days=b_duration_years * 365.25)

            paryanthars = []
            p_start_idx = DASA_LORDS.index(b_lord)

            for k in range(9):
                p_lord = DASA_LORDS[(p_start_idx + k) % 9]
                p_duration_years = (DASA_YEARS[m_lord] * DASA_YEARS[b_lord] * DASA_YEARS[p_lord]) / 14400.0
                if i == 0:
                    p_duration_years *= (remaining_first_dasa_years / DASA_YEARS[m_lord])

                p_start = b_start if k == 0 else paryanthars[-1]["end_date"]
                p_end = p_start + timedelta(days=p_duration_years * 365.25)

                paryanthars.append({
                    "paryanthar_lord": p_lord,
                    "start_date": p_start,
                    "end_date": p_end
                })

            bhuktis.append({
                "bhukti_lord": b_lord,
                "start_date": b_start,
                "end_date": b_end,
                "paryanthars": paryanthars
            })

        dasa_tree.append({
            "mahadasa_lord": m_lord,
            "start_date": m_start,
            "end_date": m_end,
            "bhuktis": bhuktis
        })

    return nakshatra_name, birth_lord, remaining_first_dasa_years, dasa_tree

def find_active_period(dasa_tree, target_date):
    for m in dasa_tree:
        if m["start_date"] <= target_date <= m["end_date"]:
            for b in m["bhuktis"]:
                if b["start_date"] <= target_date <= b["end_date"]:
                    for p in b["paryanthars"]:
                        if p["start_date"] <= target_date <= p["end_date"]:
                            return m, b, p
    return None, None, None

# ==============================================================================
# 5. DATASET LOADER & MATCHING ENGINE
# ==============================================================================
@st.cache_data
def load_dataset():
    csv_file = "panchanga_972_graha_rashi_nakshatra_pada_dataset.csv"
    if os.path.exists(csv_file):
        return pd.read_csv(csv_file)

    rows = []
    for g_code, g_name in GRAHA_KEYS.items():
        overall_pada_counter = 0
        for nak_idx, (nak_name, nak_lord) in enumerate(NAKSHATRAS_LIST):
            for pada in range(1, 5):
                rashi_idx = overall_pada_counter // 9
                rashi_name = RASHIS[rashi_idx]
                navamsha_idx = overall_pada_counter % 12
                navamsha_name = RASHIS[navamsha_idx]

                pred = f"{g_name} in {rashi_name} ({nak_name} Pada {pada}, Navamsha {navamsha_name}): Strong expression of {nak_lord} energy influencing key life outcomes."
                rows.append({
                    "Graha_Code": g_code,
                    "Graha_Name": g_name,
                    "Rashi": rashi_name,
                    "Nakshatra": nak_name,
                    "Nakshatra_Lord": nak_lord,
                    "Pada": pada,
                    "Navamsha_Sign": navamsha_name,
                    "Phalaphala_Prediction": pred
                })
                overall_pada_counter += 1
    return pd.DataFrame(rows)

df_dataset = load_dataset()

def get_matched_prediction(g_code, rashi, nakshatra, pada):
    match = df_dataset[
        (df_dataset['Graha_Code'] == g_code) &
        (df_dataset['Rashi'] == rashi) &
        (df_dataset['Nakshatra'] == nakshatra) &
        (df_dataset['Pada'] == pada)
    ]
    if not match.empty:
        return match.iloc[0]['Phalaphala_Prediction']
    return f"{GRAHA_KEYS.get(g_code, g_code)} in {rashi} under {nakshatra} Pada {pada}."

# ==============================================================================
# 6. TRANSLATION ENGINE (ENGLISH TO SINHALA & TAMIL)
# ==============================================================================
TRANSLATION_MAP = {
    "Career & Profession": {"si": "වෘත්තීය සහ රැකියාව", "ta": "தொழில் மற்றும் வேலை"},
    "Love, Marriage & Relationships": {"si": "ආදරය, විවාහය සහ සබඳතා", "ta": "காதல், திருமணம் மற்றும் உறவுகள்"},
    "Wealth & Finances": {"si": "ධනය සහ මුදල්", "ta": "செல்வம் மற்றும் நிதி"},
    "Education & Intellect": {"si": "අධ්‍යාපනය සහ බුද්ධිය", "ta": "கல்வி மற்றும் அறிவு"},
    "Health & Vitality": {"si": "සෞඛ්‍යය සහ ශක්තිය", "ta": "ஆரோக்கியம் மற்றும் நல்வாழ்வு"},
    "High vitality, leadership instinct, dignified ego, eye sensitivity.": {
        "si": "ඉහළ ජීව ශක්තිය, නායකත්ව සහජ බුද්ධිය, ගෞරවනීය අභිමානය, ඇස්වල සංවේදීතාව.",
        "ta": "உயர் உயிர்ச்சக்தி, தலைமைப் பண்பු, கண்ணியமான தன்னாளுமை, கண் உணர்திறன்."
    },
    "High intellect, professional success, administrative skill, sharp speech.": {
        "si": "ඉහළ බුද්ධිය, වෘත්තීය සාර්ථකත්වය, පාලන කුසලතාව, තියුණු කථනය.",
        "ta": "உயர் அறிவுத்திறன், தொழில்முறை வெற்றி, நிர்வாகத் திறமை, கூர்மையான பேச்சு."
    }
}

def translate_prediction(text, lang):
    if lang == "en":
        return text

    # Simple rule-based translation helper for common predictions
    for eng, trans in TRANSLATION_MAP.items():
        if eng in text and lang in trans:
            text = text.replace(eng, trans[lang])

    if lang == "si":
        prefix = "\n\n**[සිංහල පරිවර්තනය]**\nඔබගේ ජන්ම පත්‍රයට සහ සක්‍රිය දශා කාලයට අනුව, ප්‍රධාන ග්‍රහ මණ්ඩල බලපෑම සහ යෝගඵල මඟින් ඉදිරි කාලයේදී සාර්ථක ප්‍රතිඵල හිමිවේ."
        return text + prefix
    elif lang == "ta":
        prefix = "\n\n**[தமிழ் மொழிபெயர்ப்பு]**\nஉங்கள் ஜாதக கிரக நிலைகள் மற்றும் தசா புத்தி காலங்களின் அடிப்படையில், முக்கிய கிரகங்களின் சேர்க்கை உங்களுக்கு சாதகமான பலன்களை அளிக்கும்."
        return text + prefix
    return text

# ==============================================================================
# 7. SOUTH INDIAN CHART RENDERER
# ==============================================================================
def render_south_indian_chart(chart_details, chart_type="D1"):
    SOUTH_GRID_POSITIONS = {
        11: (0, 0), 0: (0, 1), 1: (0, 2), 2: (0, 3),
        3: (1, 3), 4: (2, 3), 5: (3, 3), 6: (3, 2),
        7: (3, 1), 8: (3, 0), 9: (2, 0), 10: (1, 0)
    }

    key_rashi = "rashi_idx" if chart_type == "D1" else "navamsha_rashi_idx"
    rashi_contents = {i: [] for i in range(12)}

    la_idx = chart_details["LA"][key_rashi]
    rashi_contents[la_idx].append("<span class='lagna-tag'>[LA]</span>")

    for p_code in ["SU", "MO", "MA", "ME", "JU", "VE", "SA", "RA", "KE"]:
        p_idx = chart_details[p_code][key_rashi]
        rashi_contents[p_idx].append(f"<span class='planet-tag'>{p_code}</span>")

    grid_cells = {}
    for r_idx, (row, col) in SOUTH_GRID_POSITIONS.items():
        rashi_name = RASHIS[r_idx]
        tags = " ".join(rashi_contents[r_idx])
        cell_html = f"""<div class="chart-box"><div class="rashi-label">{rashi_name}</div><div>{tags}</div></div>"""
        grid_cells[(row, col)] = cell_html

    html_out = '<div class="south-chart-container">'
    for r in range(4):
        for c in range(4):
            if (r == 1 or r == 2) and (c == 1 or c == 2):
                if r == 1 and c == 1:
                    html_out += f'<div class="chart-box-center"><div>SOUTH INDIAN<br/><b>{chart_type} CHART</b></div></div>'
            else:
                html_out += grid_cells[(r, c)]
    html_out += '</div>'
    return html_out

# ==============================================================================
# 8. STREAMLIT FRONTEND DASHBOARD
# ==============================================================================
st.markdown('<div class="main-header">🔮 Vedic Astrology Expert System & Dasa Engine</div>', unsafe_allow_html=True)
st.markdown('<div class="sub-header">South Indian Rasi & Navamsa Charts • Dasa-Bhukti Timing • Graha Descriptions • Multilingual Predictions</div>', unsafe_allow_html=True)

# --- SIDEBAR INPUTS ---
st.sidebar.header("📥 Birth Details Input")
birth_date = st.sidebar.date_input("Birth Date", datetime.date(1996, 5, 24))
time_input_str = st.sidebar.text_input("Birth Time (HH:MM in 24-hr format)", "10:30")

try:
    hrs, mins = map(int, time_input_str.split(":"))
    birth_time = datetime.time(hrs, mins)
except Exception:
    st.sidebar.error("Invalid time format. Please enter HH:MM (e.g. 14:30).")
    birth_time = datetime.time(12, 0)

st.sidebar.subheader("📍 Location Selector")
location_type = st.sidebar.radio("Location Mode", ["Popular Cities Database", "Custom Location Search"])

if location_type == "Popular Cities Database":
    selected_city = st.sidebar.selectbox("Select City", list(POPULAR_CITIES.keys()), index=0)
    lat, lon, tz_str = resolve_location(selected_city)
else:
    custom_city = st.sidebar.text_input("Enter City Name (e.g. Colombo, Sri Lanka)", "Colombo, Sri Lanka")
    lat, lon, tz_str = resolve_location(custom_city)

st.sidebar.info(f"**Coordinates:** {lat:.4f}° N, {lon:.4f}° E\n\n**Timezone:** {tz_str}")

local_tz = pytz.timezone(tz_str)
naive_dt = datetime.datetime.combine(birth_date, birth_time)
local_dt = local_tz.localize(naive_dt)
utc_dt = local_dt.astimezone(pytz.utc)

st.sidebar.caption(f"**UTC Calculated Time:** {utc_dt.strftime('%Y-%m-%d %H:%M:%S UTC')}")

# --- CALCULATIONS ---
chart = compute_positions(utc_dt, lat, lon)
moon_deg = chart["MO"]["deg"]
nak_name, birth_lord, remaining_years, dasa_tree = calculate_full_dasa_tree(naive_dt, moon_deg)

# --- TABS INTERFACE ---
tab1, tab2, tab3, tab4, tab5 = st.tabs([
    "📊 Rasi & Navamsa Charts",
    "⚡ Dasa-Bhukti Timeline",
    "🤖 Expert System Life Query",
    "📜 Graha & Yuti Descriptions",
    "📤 Data Lookups & Export"
])

# ------------------------------------------------------------------------------
# TAB 1: CHARTS VIEW
# ------------------------------------------------------------------------------
with tab1:
    col1, col2 = st.columns(2)
    with col1:
        st.subheader("Rasi Chart (D1 - Main Chart)")
        st.markdown(render_south_indian_chart(chart, "D1"), unsafe_allow_html=True)
    with col2:
        st.subheader("Navamsa Chart (D9 - Micro Chart)")
        st.markdown(render_south_indian_chart(chart, "D9"), unsafe_allow_html=True)

    st.divider()
    c1, c2, c3, c4 = st.columns(4)
    c1.metric("Ascendant (Lagna)", chart["LA"]["rashi"])
    c2.metric("Moon Sign (Rasi)", chart["MO"]["rashi"])
    c3.metric("Janma Nakshatra", f"{nak_name} (Pada {chart['MO']['pada']})")
    c4.metric("First Dasa Balance", f"{remaining_years:.2f} Yrs ({birth_lord})")

# ------------------------------------------------------------------------------
# TAB 2: DASA TIMELINE
# ------------------------------------------------------------------------------
with tab2:
    st.subheader("🌳 Vimshottari Dasa-Bhukti-Paryanthar Timeline")

    analysis_date = st.date_input("Target Analysis Date for Timing Queries", datetime.date.today())
    target_dt = datetime.datetime.combine(analysis_date, datetime.time.min)

    active_m, active_b, active_p = find_active_period(dasa_tree, target_dt)

    if active_m and active_b and active_p:
        st.success(f"**Active Period on {analysis_date.strftime('%d-%b-%Y')}:**")
        ac1, ac2, ac3 = st.columns(3)
        ac1.info(f"**Mahadasa Lord:** {active_m['mahadasa_lord']}\n\n`{active_m['start_date'].strftime('%Y-%m-%d')}` to `{active_m['end_date'].strftime('%Y-%m-%d')}`")
        ac2.success(f"**Bhukti Lord:** {active_b['bhukti_lord']}\n\n`{active_b['start_date'].strftime('%Y-%m-%d')}` to `{active_b['end_date'].strftime('%Y-%m-%d')}`")
        ac3.warning(f"**Paryanthar Lord:** {active_p['paryanthar_lord']}\n\n`{active_p['start_date'].strftime('%Y-%m-%d')}` to `{active_p['end_date'].strftime('%Y-%m-%d')}`")

    st.divider()
    st.subheader("Full 120-Year Dasa Hierarchy")
    for m in dasa_tree:
        m_label = f"Mahadasa: {m['mahadasa_lord']} ({m['start_date'].strftime('%b %Y')} - {m['end_date'].strftime('%b %Y')})"
        with st.expander(m_label):
            for b in m["bhuktis"]:
                b_label = f"--- Bhukti: {b['bhukti_lord']} ({b['start_date'].strftime('%d-%b-%Y')} to {b['end_date'].strftime('%d-%b-%Y')})"
                with st.expander(b_label):
                    p_data = [{
                        "Paryanthar Lord": p["paryanthar_lord"],
                        "Start Date": p["start_date"].strftime("%d-%b-%Y"),
                        "End Date": p["end_date"].strftime("%d-%b-%Y")
                    } for p in b["paryanthars"]]
                    st.dataframe(pd.DataFrame(p_data), use_container_width=True)

# ------------------------------------------------------------------------------
# TAB 3: EXPERT SYSTEM & TIMING QUERY
# ------------------------------------------------------------------------------
with tab3:
    st.subheader("🤖 Ask Precise Timing & Life Questions")

    query_domain = st.selectbox(
        "Select Domain or Timing Query:",
        [
            "When will I get promoted / career growth?",
            "When will I get married / find a partner?",
            "Career & Profession",
            "Love, Marriage & Relationships",
            "Wealth & Finances",
            "Education & Intellect",
            "Health & Vitality"
        ]
    )

    user_custom_q = st.text_input("Specific Custom Details / Questions:", "")

    if "prediction_lang" not in st.session_state:
        st.session_state.prediction_lang = "en"

    col_l1, col_l2, col_l3 = st.columns(3)
    if col_l1.button("Read in English 🇬🇧"):
        st.session_state.prediction_lang = "en"
    if col_l2.button("Read in Sinhala 🇱🇰 (සිංහල)"):
        st.session_state.prediction_lang = "si"
    if col_l3.button("Read in Tamil 🇱🇰 (தமிழ்)"):
        st.session_state.prediction_lang = "ta"

    if st.button("Generate Precision Prediction"):
        with st.spinner("Analyzing planetary lords, Dasa timing, and yutis..."):
            ans_bullets = []

            # Identify active periods
            curr_dt = datetime.datetime.now()
            m_act, b_act, p_act = find_active_period(dasa_tree, curr_dt)

            ans_bullets.append(f"### 🌟 Analysis for: {query_domain}")

            if "promoted" in query_domain.lower() or "career" in query_domain.lower():
                house_10_lord = GRAHA_KEYS[DASA_LORDS[(chart['LA']['rashi_idx'] + 9) % 9]]
                ans_bullets.append(f"**1. Career Timing & House Activation (10th/6th/11th Bhavas):**\nYour 10th House (Karma Bhava) lord is influenced by **{house_10_lord}**.")
                if m_act:
                    ans_bullets.append(f"**2. Active Timing Trigger:** Currently running **{m_act['mahadasa_lord']} Mahadasa** and **{b_act['bhukti_lord']} Bhukti**. Favorable promotion windows activate when Dasa lords align with 10th or 11th house lords.")
            elif "married" in query_domain.lower() or "partner" in query_domain.lower():
                house_7_lord = GRAHA_KEYS[DASA_LORDS[(chart['LA']['rashi_idx'] + 6) % 9]]
                ans_bullets.append(f"**1. Relationship Activation (7th/2nd/11th Bhavas):**\nYour 7th House (Kalatra Bhava) lord is **{house_7_lord}**.")
                if m_act:
                    ans_bullets.append(f"**2. Marriage Window:** Active sub-period under **{b_act['bhukti_lord']} Bhukti** activates relationship triggers.")
            else:
                ans_bullets.append(f"**1. General Domain Strength:** Positioned under **{chart['LA']['rashi']} Lagna** with **{chart['MO']['rashi']} Moon Sign**.")

            # Add Yuti detections
            detected_yutis = []
            rashi_groups = {}
            for p_code in ["SU", "MO", "MA", "ME", "JU", "VE", "SA", "RA", "KE"]:
                r_idx = chart[p_code]["rashi_idx"]
                rashi_groups.setdefault(r_idx, []).append(p_code)

            for r_idx, planets in rashi_groups.items():
                if len(planets) >= 2:
                    for p1, p2, name, desc, houses in YUTI_DESCRIPTIONS:
                        if p1 in planets and p2 in planets:
                            detected_yutis.append(f"• **{name} ({p1} + {p2} in {RASHIS[r_idx]}):** {desc}")

            if detected_yutis:
                ans_bullets.append("**3. Active Yuti / Combination Impacts in your Chart:**\n" + "\n".join(detected_yutis))

            full_prediction_text = "\n\n".join(ans_bullets)
            translated_text = translate_prediction(full_prediction_text, st.session_state.prediction_lang)

            st.markdown('<div class="prediction-card">', unsafe_allow_html=True)
            st.markdown(translated_text)
            st.markdown('</div>', unsafe_allow_html=True)

# ------------------------------------------------------------------------------
# TAB 4: GRAHA & YUTI DESCRIPTIONS
# ------------------------------------------------------------------------------
with tab4:
    st.subheader("📜 Individual Graha Descriptions (Bhava 1 to 12)")
    selected_g = st.selectbox("Select Planet / Graha:", list(GRAHA_DESCRIPTIONS.keys()))

    for desc in GRAHA_DESCRIPTIONS[selected_g]:
        st.write(f"• {desc}")

    st.divider()
    st.subheader("🔗 Planetary Combination (Yuti) Matrix")
    df_yuti = pd.DataFrame(YUTI_DESCRIPTIONS, columns=["Planet 1", "Planet 2", "Yoga Name", "Description", "Key Houses"])
    st.dataframe(df_yuti, use_container_width=True)

# ------------------------------------------------------------------------------
# TAB 5: DATA LOOKUPS & EXPORT
# ------------------------------------------------------------------------------
with tab5:
    st.subheader("📊 Detailed Planetary Placements & Dataset Matches")

    data_rows = []
    for p_code in ["SU", "MO", "MA", "ME", "JU", "VE", "SA", "RA", "KE"]:
        p_info = chart[p_code]
        phalaphala = generate_phalaphala(
            p_code, GRAHA_KEYS[p_code], p_info["rashi"],
            p_info["nakshatra"], p_info["pada"], p_info["navamsha_rashi"], p_info["nakshatra_lord"]
        )
        data_rows.append({
            "Code": p_code,
            "Planet": GRAHA_KEYS[p_code],
            "House": f"House {p_info['house']}",
            "Rasi (D1)": p_info["rashi"],
            "Nakshatra": p_info["nakshatra"],
            "Pada": p_info["pada"],
            "Navamsha (D9)": p_info["navamsha_rashi"],
            "Phalaphala Rule Output": phalaphala
        })

    st.dataframe(pd.DataFrame(data_rows), use_container_width=True)
````

````python
%%writefile app.py

import streamlit as st
import datetime
from datetime import timedelta
import pytz
from geopy.geocoders import Nominatim
from timezonefinder import TimezoneFinder
import pandas as pd
import numpy as np
import os
import re
import math

# Try importing swisseph for exact calculations; fallback if not available
try:
    import swisseph as swe
    HAS_SWISSEPH = True
except ImportError:
    HAS_SWISSEPH = False

# ==============================================================================
# 1. PAGE CONFIGURATION & CUSTOM STYLING
# ==============================================================================
st.set_page_config(
    page_title="Vedic Astrology Expert System & Dasa Engine",
    page_icon="🔮",
    layout="wide",
    initial_sidebar_state="expanded"
)

st.markdown("""
<style>
    .main-header {
        font-size: 2.2rem;
        color: #1E3A8A;
        text-align: center;
        font-weight: 700;
        margin-bottom: 0.2rem;
    }
    .sub-header {
        font-size: 1.0rem;
        color: #4B5563;
        text-align: center;
        margin-bottom: 1.5rem;
    }
    .south-chart-container {
        display: grid;
        grid-template-columns: repeat(4, 1fr);
        grid-template-rows: repeat(4, 110px);
        gap: 4px;
        background-color: #1E293B;
        padding: 6px;
        border-radius: 8px;
        max-width: 500px;
        margin: 0 auto;
    }
    .chart-box {
        background-color: #FFFFFF;
        border: 1px solid #CBD5E1;
        border-radius: 4px;
        padding: 4px;
        font-size: 11px;
        display: flex;
        flex-direction: column;
        justify-content: space-between;
        overflow: hidden;
    }
    .chart-box-center {
        grid-column: 2 / span 2;
        grid-row: 2 / span 2;
        background-color: #F8FAFC;
        border: 2px solid #94A3B8;
        display: flex;
        flex-direction: column;
        align-items: center;
        justify-content: center;
        font-weight: bold;
        color: #1E3A8A;
        font-size: 14px;
        text-align: center;
    }
    .rashi-label {
        font-size: 9px;
        font-weight: bold;
        color: #64748B;
        text-align: right;
    }
    .planet-tag {
        color: #1E3A8A;
        font-weight: bold;
        font-size: 11px;
    }
    .lagna-tag {
        color: #DC2626;
        font-weight: bold;
        font-size: 11px;
    }
    .prediction-card {
        background-color: #F0F9FF;
        border-left: 5px solid #0284C7;
        padding: 18px;
        border-radius: 6px;
        margin-bottom: 15px;
    }
</style>
""", unsafe_allow_html=True)

# ==============================================================================
# 2. CITIES DATABASE (SRI LANKAN & GLOBAL CITIES)
# ==============================================================================
POPULAR_CITIES = {
    # Sri Lanka
    "Colombo, Sri Lanka": {"lat": 6.9271, "lon": 79.8612, "tz": "Asia/Colombo"},
    "Kandy, Sri Lanka": {"lat": 7.2906, "lon": 80.6337, "tz": "Asia/Colombo"},
    "Jaffna, Sri Lanka": {"lat": 9.6615, "lon": 80.0255, "tz": "Asia/Colombo"},
    "Nuwara Eliya, Sri Lanka": {"lat": 6.9497, "lon": 80.7891, "tz": "Asia/Colombo"},
    "Galle, Sri Lanka": {"lat": 6.0535, "lon": 80.2210, "tz": "Asia/Colombo"},
    "Kurunegala, Sri Lanka": {"lat": 7.4863, "lon": 80.3623, "tz": "Asia/Colombo"},
    "Batticaloa, Sri Lanka": {"lat": 7.7310, "lon": 81.6747, "tz": "Asia/Colombo"},
    "Trincomalee, Sri Lanka": {"lat": 8.5874, "lon": 81.2152, "tz": "Asia/Colombo"},
    "Ampara, Sri Lanka": {"lat": 7.2912, "lon": 81.6724, "tz": "Asia/Colombo"},
    "Badulla, Sri Lanka": {"lat": 6.9934, "lon": 81.0550, "tz": "Asia/Colombo"},
    "Vavuniya, Sri Lanka": {"lat": 8.7514, "lon": 80.4971, "tz": "Asia/Colombo"},
    "Mannar, Sri Lanka": {"lat": 8.9810, "lon": 79.9044, "tz": "Asia/Colombo"},
    "Hatton, Sri Lanka": {"lat": 6.8833, "lon": 80.5961, "tz": "Asia/Colombo"},
    "Ratnapura, Sri Lanka": {"lat": 6.6828, "lon": 80.4016, "tz": "Asia/Colombo"},
    "Matara, Sri Lanka": {"lat": 5.9549, "lon": 80.5550, "tz": "Asia/Colombo"},
    "Anuradhapura, Sri Lanka": {"lat": 8.3114, "lon": 80.4037, "tz": "Asia/Colombo"},
    "Polonnaruwa, Sri Lanka": {"lat": 7.9403, "lon": 81.0188, "tz": "Asia/Colombo"},
    "Hambantota, Sri Lanka": {"lat": 6.1241, "lon": 81.1185, "tz": "Asia/Colombo"},
    "Puttalam, Sri Lanka": {"lat": 8.0362, "lon": 79.8283, "tz": "Asia/Colombo"},

    # India & Global
    "New Delhi, India": {"lat": 28.6139, "lon": 77.2090, "tz": "Asia/Kolkata"},
    "Chennai, India": {"lat": 13.0827, "lon": 80.2707, "tz": "Asia/Kolkata"},
    "Mumbai, India": {"lat": 19.0760, "lon": 72.8777, "tz": "Asia/Kolkata"},
    "London, United Kingdom": {"lat": 51.5074, "lon": -0.1278, "tz": "Europe/London"},
    "New York, USA": {"lat": 40.7128, "lon": -74.0060, "tz": "America/New_York"},
    "Singapore": {"lat": 1.3521, "lon": 103.8198, "tz": "Asia/Singapore"},
    "Dubai, UAE": {"lat": 25.2048, "lon": 55.2708, "tz": "Asia/Dubai"},
    "Sydney, Australia": {"lat": -33.8688, "lon": 151.2093, "tz": "Australia/Sydney"}
}

def resolve_location(location_query):
    if location_query in POPULAR_CITIES:
        data = POPULAR_CITIES[location_query]
        return data["lat"], data["lon"], data["tz"]
    try:
        geolocator = Nominatim(user_agent="vedic_astrology_app_2026")
        loc = geolocator.geocode(location_query, timeout=5)
        if loc:
            tf = TimezoneFinder()
            tz_str = tf.timezone_at(lat=loc.latitude, lng=loc.longitude)
            return loc.latitude, loc.longitude, tz_str or "UTC"
    except Exception:
        pass
    return 6.9271, 79.8612, "Asia/Colombo"

# ==============================================================================
# 3. KNOWLEDGE BASES & HOUSE LORD MAPS
# ==============================================================================
RASHIS = [
    "Mesham", "Rishabham", "Mithunam", "Karkataka",
    "Simham", "Kanya", "Thula", "Vrischikam",
    "Dhanus", "Makaram", "Kumbham", "Meenam"
]

# Rasi House Rulers (0: Mesham -> MA, 1: Rishabham -> VE, etc.)
RASI_LORDS_CODES = ["MA", "VE", "ME", "MO", "SU", "ME", "VE", "MA", "JU", "SA", "SA", "JU"]

GRAHA_KEYS = {
    "SU": "Sun", "MO": "Moon", "MA": "Mars", "ME": "Mercury",
    "JU": "Jupiter", "VE": "Venus", "SA": "Saturn", "RA": "Rahu", "KE": "Ketu"
}

def get_house_lord(lagna_rashi_idx, house_num):
    """
    Returns (lord_code, lord_name) for any given house_num (1 to 12) relative to Lagna.
    """
    target_rashi_idx = (lagna_rashi_idx + (house_num - 1)) % 12
    lord_code = RASI_LORDS_CODES[target_rashi_idx]
    return lord_code, GRAHA_KEYS[lord_code]

GRAHA_DESCRIPTIONS = {
    "SU": [
        "B1: Strong vitality, leadership instinct, dignified ego, eye sensitivity.",
        "B2: Wealth via authority/father, direct speech, potential family friction.",
        "B3: High courage, victorious over rivals, skilled hands, energetic siblings.",
        "B4: Property interest, high self-respect, emotional pride, maternal distance.",
        "B5: Sharp intellect, strong creative drive, high ambition for offspring.",
        "B6: High resistance, dominant over opponents, legal success, heat disorders.",
        "B7: Strong-willed partner, high standard in diplomacy, ego trade-offs.",
        "B8: Interest in occult/research, sudden transformation, vitality care needed.",
        "B9: Righteous nature, connection to government/law, strong moral compass.",
        "B10: Directional Strength (Dig Bala), fast professional rise, executive authority.",
        "B11: High gains through government/influential contacts, stable prosperity.",
        "B12: Interest in solitary pursuits, foreign expenditure, spiritual detachment."
    ],
    "MO": [
        "B1: Sensitive demeanor, public charm, expressive face, fluctuating moods.",
        "B2: Wealth through public ventures, sweet speech, family emotional attachment.",
        "B3: Literary interest, communicative talent, frequent short travel.",
        "B4: Directional Strength (Dig Bala), peace of mind, maternal blessings, comfort.",
        "B5: Intuitive intelligence, poetic mind, strong attachment to children.",
        "B6: Emotional fluctuations under stress, service nature, care with digestive health.",
        "B7: Public attraction, empathetic partner, emotionally responsive relationships.",
        "B8: Unconscious insights, interest in secrets, sudden emotional shifts.",
        "B9: Pilgrimages, philosophical mind, fortune through grace and travel.",
        "B10: Career with public visibility, caregiving/hospitality talent, variable fame.",
        "B11: Wide social network, fluctuating income streams, helpful female friends.",
        "B12: Active imagination, need for solitude, interest in foreign lands/retreats."
    ],
    "MA": [
        "B1: High physical vigor, impulsive drive, leadership ability, physical marks.",
        "B2: Bold speech, financial assertiveness, direct approach in family affairs.",
        "B3: Immense courage, mechanical prowess, protective of siblings, sports ability.",
        "B4: Property ownership, technical mind, energetic home environment, assertiveness.",
        "B5: Sharp tactical intelligence, competitive spirit, bold investment style.",
        "B6: Shatters opposition (Shatru Hanta), high physical stamina, legal triumph.",
        "B7: Dynamic partner, high passion, requires patience in relationships.",
        "B8: Interest in research/engineering, transformation through intense experiences.",
        "B9: Passionate belief system, energetic defender of truth, independent thinker.",
        "B10: Directional Strength (Dig Bala), top performance in engineering/management.",
        "B11: Large financial gains through effort, competitive edge, influential allies.",
        "B12: High energy spent in private pursuits, interest in foreign defense/travel."
    ],
    "ME": [
        "B1: Directional Strength (Dig Bala), sharp intellect, youthful look, eloquent speech.",
        "B2: Financial literacy, skilled in negotiation, family business aptitude.",
        "B3: Outstanding writing/media talent, fast learning curve, active networking.",
        "B4: Intellectual home environment, interest in education/real estate analysis.",
        "B5: Analytical capacity, interest in astrology/mathematics, sharp memory.",
        "B6: Analytical problem solving, debate champion, careful health tracking.",
        "B7: Intellectual companion, diplomatic skill, youthful business partner.",
        "B8: Deep analytical research, interest in codes/cryptography, secret knowledge.",
        "B9: Higher learning, publishing aptitude, philosophical writing, broad outlook.",
        "B10: Corporate management, administrative talent, multi-tasking career.",
        "B11: Diverse income sources, strategic networking, commercial success.",
        "B12: Research in solitude, expenditure on education, subtle analytical mind."
    ],
    "JU": [
        "B1: Directional Strength (Dig Bala), immense wisdom, moral clarity, sound health.",
        "B2: Wise speech, financial abundance, family prosperity, ethical values.",
        "B3: Philosophical advisor, constructive communication, supportive brothers.",
        "B4: Domestic contentment, large house, maternal blessings, vehicles.",
        "B5: Exceptional intelligence (Prajna), spiritual merit, honored children.",
        "B6: Overcomes obstacles through wisdom, legal victory, service mentality.",
        "B7: Devoted/wise partner, high social reputation, balanced contracts.",
        "B8: Unlocks hidden wisdom, long life, seamless legacy asset inheritance.",
        "B9: Natural house of fortune (Bhagya), higher teacher, spiritual devotion.",
        "B10: Highly respected career, advisory roles, judicial/executive success.",
        "B11: Steady accumulation of wealth, elite mentorship, noble aspirations.",
        "B12: Higher spiritual liberation (Moksha), charitable nature, foreign peace."
    ],
    "VE": [
        "B1: Charming personality, refined taste, artistic presence, graceful health.",
        "B2: Luxurious speech, wealth through arts/media, pleasant family life.",
        "B3: Artistic talent in music/writing, sweet relationship with siblings.",
        "B4: Directional Strength (Dig Bala), luxury vehicles, beautiful home, comfort.",
        "B5: Artistic intelligence, romance, skill in design and entertainment.",
        "B6: Gracious handling of conflict, diplomatic resolution of disputes.",
        "B7: Romantic/refined spouse, harmonious business partnerships, public success.",
        "B8: Legacy wealth, smooth joint finances, interest in esoteric arts.",
        "B9: Appreciation for cultural arts, fortune through partnerships/travel.",
        "B10: Career in luxury/arts/diplomacy, high social prestige, pleasant work environment.",
        "B11: Wealth from artistic/female networks, fulfillment of refined desires.",
        "B12: Exaltation house effect (in Pisces), high comfort, expenditure on art/luxury."
    ],
    "SA": [
        "B1: Serious outlook, disciplined nature, slow & steady success, endurance.",
        "B2: Conservative spending, measured speech, delayed accumulation of wealth.",
        "B3: Great patience, persistent effort, strong hands, strategic thinking.",
        "B4: Traditional home, duty toward mother, property through hard work.",
        "B5: Deep focus, structured study, serious attitude toward responsibilities.",
        "B6: Dominates long-term enemies, exceptional work ethic, stamina in crises.",
        "B7: Directional Strength (Dig Bala), mature/loyal partner, enduring contracts.",
        "B8: Exceptional longevity, interest in history/geology, deep endurance.",
        "B9: Respect for traditional wisdom, methodical path to fortune, discipline.",
        "B10: Natural house of career, high organizational power, durable leadership.",
        "B11: Massive wealth built over time, loyal long-term friends, stable gains.",
        "B12: Solitary discipline, low material desires, interest in spiritual austerity."
    ],
    "RA": [
        "B1: Unconventional magnetic personality, innovative drive, distinct worldview.",
        "B2: Unique wealth streams, unconventional speech, ambition for resources.",
        "B3: Extraordinary courage, mastery over digital/foreign communications.",
        "B4: Non-traditional living arrangements, foreign real estate interests.",
        "B5: Unconventional intellect, original creative ideas, speculative skill.",
        "B6: Unbeatable against opponents, triumphs in complex legal battles.",
        "B7: Unique/foreign partner, breakthrough approach to partnerships.",
        "B8: Deep psychological insight, sudden discoveries, esoteric interests.",
        "B9: Questions traditions, seeks knowledge from foreign/unconventional paths.",
        "B10: Sudden rise to prominent career status, media appeal, disruptive innovation.",
        "B11: Multiple massive income sources, vast global network, ambitious goals.",
        "B12: Foreign residency, interest in dreams, private spiritual exploratory mind."
    ],
    "KE": [
        "B1: Spiritual aura, intuitive nature, detached identity, self-inquiry.",
        "B2: Minimalist speech, detachment from material hoarding, intuitive values.",
        "B3: Subtle non-verbal skills, courage in quiet ways, deep internal focus.",
        "B4: Simple home preference, intuitive connection to roots, emotional detachment.",
        "B5: Deep abstract intelligence, karmic insights, unique spiritual mantras.",
        "B6: Instinctive crisis management, detachment from conflicts.",
        "B7: Spiritual/introverted partner, search for deeper soul connection.",
        "B8: Esoteric mastery, mastery over occult sciences, transformational liberation.",
        "B9: Direct spiritual insight, non-materialistic philosophy, pilgrimage.",
        "B10: Unconventional career path, indifference to titles, specialized skill.",
        "B11: Selective circle of friends, spiritual gains over material pursuits.",
        "B12: Natural house of Moksha for Ketu, ultimate spiritual liberation, deep meditation."
    ]
}

YUTI_DESCRIPTIONS = [
    ("SU", "MO", "Amavasya Yuti", "High internal focus, strong alignment of soul & mind, introverted drive.", ["B1", "B5", "B9", "B10"]),
    ("SU", "MA", "Agni-Tejas Yoga", "Immense energy, physical vitality, executive leadership, strong competitive edge.", ["B3", "B6", "B10"]),
    ("SU", "ME", "Budhaditya Yoga", "High intellect, professional success, administrative skill, sharp speech.", ["B1", "B4", "B7", "B10", "B11"]),
    ("SU", "JU", "Guru-Aditya Yoga", "Moral authority, legal wisdom, favored by elders/government, high ethics.", ["B1", "B5", "B9", "B10"]),
    ("SU", "VE", "Venus Combustion / Yuti", "Artistic drive, passionate nature, high luxury ambition, requires relationship care.", ["B1", "B5", "B10"]),
    ("SU", "SA", "Pitri Dosha / Sun-Saturn", "Ego friction with authority, immense discipline under pressure, hard-earned triumph.", ["B3", "B6", "B10", "B11"]),
    ("SU", "RA", "Grahan Yoga (Solar)", "High ambition, sudden political or public rise, quest for unique recognition.", ["B3", "B6", "B10", "B11"]),
    ("SU", "KE", "Solar Moksha Yuti", "Spiritual detachment from fame, deep introspection, self-inquiry, research capacity.", ["B9", "B12"]),
    ("MO", "MA", "Chandra-Mangala Yoga", "Dynamic financial drive, success in real estate/trade, emotional intensity.", ["B2", "B9", "B10", "B11"]),
    ("MO", "ME", "Chandra-Budha Yoga", "Quick wit, commercial acumen, literary skill, highly adaptable intelligence.", ["B1", "B2", "B5", "B10"]),
    ("MO", "JU", "Gaja-Kesari Yoga", "Lasting reputation, moral integrity, financial stability, high public protection.", ["B1", "B4", "B7", "B9", "B10"]),
    ("MO", "VE", "Chandra-Shukra Yoga", "Refined artistic tastes, personal charm, success in luxury/hospitality, peace.", ["B1", "B4", "B7", "B11"]),
    ("MO", "SA", "Vish Yoga / Punarphoo", "Emotional discipline, mental endurance, inclination toward solitary work.", ["B3", "B6", "B10", "B11"]),
    ("MO", "RA", "Kemu-Rahu Yuti", "High imagination, psychological depth, intense intuition, creative innovation.", ["B3", "B6", "B10", "B11"]),
    ("MO", "KE", "Kemu-Ketu Yuti", "Deep emotional detachment, psychic sensitivity, inclination toward meditation.", ["B8", "B9", "B12"]),
    ("MA", "ME", "Krodha / Engineering Yuti", "Sharp argumentative skill, technical talent, quick reflexes, debate mastery.", ["B3", "B6", "B10"]),
    ("MA", "JU", "Guru-Mangala Yoga", "Righteous action, legal victory, executive courage, protective instinct.", ["B1", "B3", "B5", "B9", "B10"]),
    ("MA", "VE", "Kama Yoga / Passion Yuti", "High creative energy, magnetic attraction, passion for arts/design, charm.", ["B1", "B5", "B7", "B10"]),
    ("MA", "SA", "Agni-Maruta / Collision", "Tremendous stamina, mechanical prowess, persistence through obstacles.", ["B3", "B6", "B10", "B11"]),
    ("MA", "RA", "Angarak Yoga", "High impulse, breakthrough explosive energy, success in high-risk engineering/defense.", ["B3", "B6", "B10", "B11"]),
    ("MA", "KE", "Kulkarna Yuti", "Precision skill, surgical efficiency, deep technical focus, sudden action.", ["B3", "B6", "B11"]),
    ("ME", "JU", "Saraswati Yoga Component", "Scholarly wisdom, financial acumen, teaching ability, literary mastery.", ["B1", "B2", "B5", "B9", "B10"]),
    ("ME", "VE", "Mahalaxmi / Design Yoga", "Elegance in speech, artistic & literary success, diplomatic charm.", ["B1", "B2", "B4", "B5", "B10"]),
    ("ME", "SA", "Nipuna / Structured Mind", "Methodical research, scientific orientation, organized thinking, endurance.", ["B2", "B3", "B6", "B10", "B11"]),
    ("ME", "RA", "Research / Technical Yuti", "Breakthrough technological talent, foreign trade skill, innovative media skill.", ["B3", "B6", "B10", "B11"]),
    ("ME", "KE", "Codes / Micro-Intellect Yuti", "Abstract coding talent, intuitive logic, deep research capability.", ["B5", "B8", "B12"]),
    ("JU", "VE", "Dvi-Guru Yoga", "Combined wisdom of laws & diplomacy, vast wealth potential, balanced guidance.", ["B1", "B4", "B5", "B9", "B11"]),
    ("JU", "SA", "Brahma / Saturn-Jupiter", "Balanced judgment, long-term wealth stability, philosophical maturity.", ["B1", "B5", "B9", "B10", "B11"]),
    ("JU", "RA", "Guru-Chandal Yoga", "Unconventional wisdom, questioning dogma, global philosophical reach.", ["B3", "B6", "B10", "B11"]),
    ("JU", "KE", "Gyan Yoga", "Highest spiritual knowledge, natural intuition, detachment from material vanity.", ["B5", "B9", "B12"]),
    ("VE", "SA", "Kalanidhi Component", "Durable artistic skill, success in fashion/architecture industry, steady finance.", ["B2", "B3", "B7", "B10", "B11"]),
    ("VE", "RA", "Asura / Luxury Expansion", "Unconventional aesthetic, international fame in media/arts, high luxury ambition.", ["B1", "B5", "B7", "B10", "B11"]),
    ("VE", "KE", "Detached Aesthetics", "Subtle artistic style, spiritual love, non-materialistic view of relationships.", ["B9", "B12"]),
    ("SA", "RA", "Shani-Rahu / Karmic Yuti", "Technical mastery, heavy structural discipline, resilience under extreme conditions.", ["B3", "B6", "B10", "B11"]),
    ("SA", "KE", "Shani-Ketu / Austerity", "Strict discipline, minimalist lifestyle, deep research, spiritual endurance.", ["B3", "B9", "B12"])
]

def generate_phalaphala(graha_code, graha_name, rashi, nakshatra, pada, navamsha, nak_lord):
    if graha_code == "SU":
        if rashi == "Mesham" and nakshatra == "Krittika" and pada == 1:
            return "Deep Exalted Peak (10 deg), supreme authority, legal victory, regal honor, high vitality."
        elif rashi == "Thula":
            return "Debilitated placement, requires ego restraint, diplomacy challenges, public service focus."
        elif rashi == "Simham":
            return "Own / Moolatrikona placement, natural executive leadership, high vitality, commanding presence."
        else:
            return f"Leadership and willpower expressed through {rashi} traits, guided by {nak_lord} energy in {navamsha} Navamsha."

    elif graha_code == "MO":
        if rashi == "Rishabham":
            return "Exalted placement, mental peace, emotional stability, wealth accumulation, luxury."
        elif rashi == "Vrischikam":
            return "Debilitated placement, emotional intensity, psychological depth, transformation."
        elif rashi == "Karkataka":
            return "Own sign placement, deep caretaking, high intuition, strong maternal connection."
        else:
            return f"Emotional expression and public mind tuned to {rashi}, shaped by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "MA":
        if rashi == "Makaram":
            return "Exalted placement, high strategic endurance, engineering skill, disciplined courage."
        elif rashi == "Karkataka":
            return "Debilitated placement, emotional reactivity, protective instincts, needs energy canalization."
        elif rashi in ["Mesham", "Vrischikam"]:
            return "Own sign placement, bold drive, pioneering action, high competitive stamina."
        else:
            return f"Dynamic drive and assertiveness channeled through {rashi}, influenced by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "ME":
        if rashi == "Kanya":
            return "Exalted / Moolatrikona placement, sharp analytical mind, technical perfectionism, business acumen."
        elif rashi == "Meenam":
            return "Debilitated placement, abstract intuitive thinking, creative imagination, poetic skill."
        elif rashi == "Mithunam":
            return "Own sign placement, high eloquence, commercial adaptability, media dexterity."
        else:
            return f"Intellectual and commercial skills expressed through {rashi}, under {nak_lord} energy in {navamsha} Navamsha."

    elif graha_code == "JU":
        if rashi == "Karkataka":
            return "Exalted placement, benevolent wisdom, high moral standing, expansive fortune, protective advisor."
        elif rashi == "Makaram":
            return "Debilitated placement, pragmatic philosophy, material focus, conservative ethics."
        elif rashi in ["Dhanus", "Meenam"]:
            return "Own sign placement, high spiritual learning, advisory authority, philosophical depth."
        else:
            return f"Wisdom, growth, and moral guidance manifested through {rashi}, guided by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "VE":
        if rashi == "Meenam":
            return "Exalted placement, unconditional devotion, high artistic elegance, refined luxury."
        elif rashi == "Kanya":
            return "Debilitated placement, meticulous aesthetic taste, analytical approach to relationships."
        elif rashi in ["Rishabham", "Thula"]:
            return "Own sign placement, personal charm, artistic talent, financial diplomacy, harmony."
        else:
            return f"Creative, romantic, and diplomatic qualities refined in {rashi}, shaped by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "SA":
        if rashi == "Thula":
            return "Exalted placement, highest justice, diplomatic endurance, long-term organizational success."
        elif rashi == "Mesham":
            return "Debilitated placement, lessons in patience, impulsive resistance, hard-earned growth."
        elif rashi in ["Makaram", "Kumbham"]:
            return "Own / Moolatrikona placement, deep discipline, mass leadership, technical stamina."
        else:
            return f"Structured discipline, duty, and endurance executed through {rashi}, influenced by {nak_lord} in {navamsha} Navamsha."

    elif graha_code == "RA":
        return f"Amplified material drive and innovative ambition in {rashi}, seeking expansion in {nakshatra} ({navamsha} Navamsha)."

    elif graha_code == "KE":
        return f"Spiritual detachment, intuitive research, and karmic insights in {rashi}, rooted in {nakshatra} ({navamsha} Navamsha)."

# ==============================================================================
# 4. VEDIC ASTROLOGY CALCULATIONS & DASA TIMING ENGINE
# ==============================================================================
NAKSHATRAS_LIST = [
    ("Ashwini", "Ketu"), ("Bharani", "Venus"), ("Krittika", "Sun"),
    ("Rohini", "Moon"), ("Mrigashira", "Mars"), ("Ardra", "Rahu"),
    ("Punarvasu", "Jupiter"), ("Pushya", "Saturn"), ("Ashlesha", "Mercury"),
    ("Magha", "Ketu"), ("Purva Phalguni", "Venus"), ("Uttara Phalguni", "Sun"),
    ("Hasta", "Moon"), ("Chitra", "Mars"), ("Swati", "Rahu"),
    ("Vishakha", "Jupiter"), ("Anuradha", "Saturn"), ("Jyeshta", "Mercury"),
    ("Mula", "Ketu"), ("Purvashadha", "Venus"), ("Uttarashadha", "Sun"),
    ("Shravana", "Moon"), ("Dhanishta", "Mars"), ("Shatabhishak", "Rahu"),
    ("Purva Bhadra", "Jupiter"), ("Uttara Bhadra", "Saturn"), ("Revati", "Mercury")
]

DASA_ORDER = [
    ("Ketu", 7), ("Venus", 20), ("Sun", 6), ("Moon", 10),
    ("Mars", 7), ("Rahu", 18), ("Jupiter", 16), ("Saturn", 19), ("Mercury", 17)
]
DASA_YEARS = dict(DASA_ORDER)
DASA_LORDS = [lord for lord, _ in DASA_ORDER]

def calculate_julian_day(dt):
    year, month, day = dt.year, dt.month, dt.day
    hour = dt.hour + dt.minute / 60.0 + dt.second / 3600.0
    if month <= 2:
        year -= 1
        month += 12
    A = math.floor(year / 100)
    B = 2 - A + math.floor(A / 4)
    jd = math.floor(365.25 * (year + 4716)) + math.floor(30.6001 * (month + 1)) + day + B - 1524.5 + (hour / 24.0)
    return jd

def compute_positions(birth_dt_utc, lat, lon):
    jd = calculate_julian_day(birth_dt_utc)

    if HAS_SWISSEPH:
        swe.set_sid_mode(swe.SIDM_LAHIRI)
        cusps, ascmc = swe.houses_ex(jd, lat, lon, b'P', swe.FLG_SIDEREAL)
        lagna_deg = ascmc[0] % 360.0

        planet_map = {
            "SU": swe.SUN, "MO": swe.MOON, "MA": swe.MARS, "ME": swe.MERCURY,
            "JU": swe.JUPITER, "VE": swe.VENUS, "SA": swe.SATURN, "RA": swe.MEAN_NODE
        }
        positions = {}
        for p_code, p_id in planet_map.items():
            res, flag = swe.calc_ut(jd, p_id, swe.FLG_SIDEREAL)
            positions[p_code] = res[0] % 360.0
        positions["KE"] = (positions["RA"] + 180.0) % 360.0
    else:
        # Fallback Lahiri calculation
        T = (jd - 2451545.0) / 36525.0
        ayanamsa = 23.85 + 1.396 * T
        days_since_j2000 = jd - 2451545.0

        positions = {
            "SU": (280.46 + 0.9856474 * days_since_j2000 - ayanamsa) % 360.0,
            "MO": (218.32 + 13.176396 * days_since_j2000 - ayanamsa) % 360.0,
            "MA": (355.43 + 0.5240330 * days_since_j2000 - ayanamsa) % 360.0,
            "ME": (252.25 + 1.6021300 * days_since_j2000 - ayanamsa) % 360.0,
            "JU": (34.35 + 0.0830910 * days_since_j2000 - ayanamsa) % 360.0,
            "VE": (181.98 + 1.6021300 * days_since_j2000 - ayanamsa) % 360.0,
            "SA": (50.08 + 0.0334596 * days_since_j2000 - ayanamsa) % 360.0,
            "RA": (125.04 - 0.0529539 * days_since_j2000 - ayanamsa) % 360.0
        }
        positions["KE"] = (positions["RA"] + 180.0) % 360.0
        lst = (100.46 + 0.985647366 * days_since_j2000 + lon + (birth_dt_utc.hour + birth_dt_utc.minute/60.0)*15) % 360.0
        lagna_deg = (lst - ayanamsa) % 360.0

    chart_details = {}
    lagna_rashi_idx = int(lagna_deg // 30)
    chart_details["LA"] = {
        "deg": lagna_deg,
        "rashi_idx": lagna_rashi_idx,
        "rashi": RASHIS[lagna_rashi_idx],
        "navamsha_rashi_idx": int((lagna_deg * 9 / 30) % 12),
        "navamsha_rashi": RASHIS[int((lagna_deg * 9 / 30) % 12)]
    }

    for p_code, deg in positions.items():
        rashi_idx = int(deg // 30)
        overall_pada_idx = int(deg // (30 / 9))
        nak_idx = overall_pada_idx // 4
        pada = (overall_pada_idx % 4) + 1
        navamsha_rashi_idx = overall_pada_idx % 12
        house_num = ((rashi_idx - lagna_rashi_idx) % 12) + 1

        chart_details[p_code] = {
            "deg": deg,
            "rashi_idx": rashi_idx,
            "rashi": RASHIS[rashi_idx],
            "nakshatra": NAKSHATRAS_LIST[nak_idx % 27][0],
            "nakshatra_lord": NAKSHATRAS_LIST[nak_idx % 27][1],
            "pada": pada,
            "navamsha_rashi_idx": navamsha_rashi_idx,
            "navamsha_rashi": RASHIS[navamsha_rashi_idx],
            "house": house_num,
            "overall_pada_idx": overall_pada_idx
        }

    return chart_details

def calculate_full_dasa_tree(dob, moon_degree):
    arc_per_nakshatra = 13.333333333333334
    nakshatra_idx = int(moon_degree // arc_per_nakshatra) % 27
    degree_in_nakshatra = moon_degree % arc_per_nakshatra
    traversed_fraction = degree_in_nakshatra / arc_per_nakshatra

    nakshatra_name = NAKSHATRAS_LIST[nakshatra_idx][0]
    birth_lord = DASA_LORDS[nakshatra_idx % 9]
    remaining_first_dasa_years = DASA_YEARS[birth_lord] * (1.0 - traversed_fraction)

    start_index = DASA_LORDS.index(birth_lord)
    current_date = dob
    dasa_tree = []

    for i in range(9):
        m_lord = DASA_LORDS[(start_index + i) % 9]
        m_duration_years = remaining_first_dasa_years if i == 0 else DASA_YEARS[m_lord]
        m_start = current_date
        m_end = m_start + timedelta(days=m_duration_years * 365.25)
        current_date = m_end

        bhuktis = []
        b_start_idx = DASA_LORDS.index(m_lord)

        for j in range(9):
            b_lord = DASA_LORDS[(b_start_idx + j) % 9]
            b_duration_years = (DASA_YEARS[m_lord] * DASA_YEARS[b_lord]) / 120.0
            if i == 0:
                b_duration_years *= (remaining_first_dasa_years / DASA_YEARS[m_lord])

            b_start = m_start if j == 0 else bhuktis[-1]["end_date"]
            b_end = b_start + timedelta(days=b_duration_years * 365.25)

            paryanthars = []
            p_start_idx = DASA_LORDS.index(b_lord)

            for k in range(9):
                p_lord = DASA_LORDS[(p_start_idx + k) % 9]
                p_duration_years = (DASA_YEARS[m_lord] * DASA_YEARS[b_lord] * DASA_YEARS[p_lord]) / 14400.0
                if i == 0:
                    p_duration_years *= (remaining_first_dasa_years / DASA_YEARS[m_lord])

                p_start = b_start if k == 0 else paryanthars[-1]["end_date"]
                p_end = p_start + timedelta(days=p_duration_years * 365.25)

                paryanthars.append({
                    "paryanthar_lord": p_lord,
                    "start_date": p_start,
                    "end_date": p_end
                })

            bhuktis.append({
                "bhukti_lord": b_lord,
                "start_date": b_start,
                "end_date": b_end,
                "paryanthars": paryanthars
            })

        dasa_tree.append({
            "mahadasa_lord": m_lord,
            "start_date": m_start,
            "end_date": m_end,
            "bhuktis": bhuktis
        })

    return nakshatra_name, birth_lord, remaining_first_dasa_years, dasa_tree

def find_active_period(dasa_tree, target_date):
    for m in dasa_tree:
        if m["start_date"] <= target_date <= m["end_date"]:
            for b in m["bhuktis"]:
                if b["start_date"] <= target_date <= b["end_date"]:
                    for p in b["paryanthars"]:
                        if p["start_date"] <= target_date <= p["end_date"]:
                            return m, b, p
    return None, None, None

# ==============================================================================
# 5. DATASET LOADER & MATCHING ENGINE
# ==============================================================================
@st.cache_data
def load_dataset():
    csv_file = "panchanga_972_graha_rashi_nakshatra_pada_dataset.csv"
    if os.path.exists(csv_file):
        return pd.read_csv(csv_file)

    rows = []
    for g_code, g_name in GRAHA_KEYS.items():
        overall_pada_counter = 0
        for nak_idx, (nak_name, nak_lord) in enumerate(NAKSHATRAS_LIST):
            for pada in range(1, 5):
                rashi_idx = overall_pada_counter // 9
                rashi_name = RASHIS[rashi_idx]
                navamsha_idx = overall_pada_counter % 12
                navamsha_name = RASHIS[navamsha_idx]

                pred = f"{g_name} in {rashi_name} ({nak_name} Pada {pada}, Navamsha {navamsha_name}): Strong expression of {nak_lord} energy influencing key life outcomes."
                rows.append({
                    "Graha_Code": g_code,
                    "Graha_Name": g_name,
                    "Rashi": rashi_name,
                    "Nakshatra": nak_name,
                    "Nakshatra_Lord": nak_lord,
                    "Pada": pada,
                    "Navamsha_Sign": navamsha_name,
                    "Phalaphala_Prediction": pred
                })
                overall_pada_counter += 1
    return pd.DataFrame(rows)

df_dataset = load_dataset()

def get_matched_prediction(g_code, rashi, nakshatra, pada):
    match = df_dataset[
        (df_dataset['Graha_Code'] == g_code) &
        (df_dataset['Rashi'] == rashi) &
        (df_dataset['Nakshatra'] == nakshatra) &
        (df_dataset['Pada'] == pada)
    ]
    if not match.empty:
        return match.iloc[0]['Phalaphala_Prediction']
    return f"{GRAHA_KEYS.get(g_code, g_code)} in {rashi} under {nakshatra} Pada {pada}."

# ==============================================================================
# 6. TRANSLATION ENGINE (ENGLISH TO SINHALA & TAMIL)
# ==============================================================================
TRANSLATION_MAP = {
    "Career & Profession": {"si": "වෘත්තීය සහ රැකියාව", "ta": "தொழில் மற்றும் வேலை"},
    "Love, Marriage & Relationships": {"si": "ආදරය, විවාහය සහ සබඳතා", "ta": "காதல், திருமணம் மற்றும் உறவுகள்"},
    "Wealth & Finances": {"si": "ධනය සහ මුදල්", "ta": "செல்வம் மற்றும் நிதி"},
    "Education & Intellect": {"si": "අධ්‍යාපනය සහ බුද්ධිය", "ta": "கல்වි மற்றும் அறிவு"},
    "Health & Vitality": {"si": "සෞඛ්‍යය සහ ශක්තිය", "ta": "ஆரோக்கியம் மற்றும் நල්வாழ்வு"}
}

def translate_prediction(text, lang):
    if lang == "en":
        return text

    for eng, trans in TRANSLATION_MAP.items():
        if eng in text and lang in trans:
            text = text.replace(eng, trans[lang])

    if lang == "si":
        prefix = "\n\n**[සිංහල පරිවර්තනය]**\nඔබගේ ජන්ම පත්‍රයට සහ සක්‍රිය දශා කාලයට අනුව, ප්‍රධාන ග්‍රහ මණ්ඩල බලපෑම සහ යෝගඵල මඟින් ඉදිරි කාලයේදී සාර්ථක ප්‍රතිඵල හිමිවේ."
        return text + prefix
    elif lang == "ta":
        prefix = "\n\n**[தமிழ் மொழிபெயர்ப்பு]**\nஉங்கள் ஜாதக கிரக நிலைகள் மற்றும் தசா புத்தி காலங்களின் அடிப்படையில், முக்கிய கிரகங்களின் சேர்க்கை உங்களுக்கு சாதகமான பலன்களை அளிக்கும்."
        return text + prefix
    return text

# ==============================================================================
# 7. SOUTH INDIAN CHART RENDERER
# ==============================================================================
def render_south_indian_chart(chart_details, chart_type="D1"):
    SOUTH_GRID_POSITIONS = {
        11: (0, 0), 0: (0, 1), 1: (0, 2), 2: (0, 3),
        3: (1, 3), 4: (2, 3), 5: (3, 3), 6: (3, 2),
        7: (3, 1), 8: (3, 0), 9: (2, 0), 10: (1, 0)
    }

    key_rashi = "rashi_idx" if chart_type == "D1" else "navamsha_rashi_idx"
    rashi_contents = {i: [] for i in range(12)}

    la_idx = chart_details["LA"][key_rashi]
    rashi_contents[la_idx].append("<span class='lagna-tag'>[LA]</span>")

    for p_code in ["SU", "MO", "MA", "ME", "JU", "VE", "SA", "RA", "KE"]:
        p_idx = chart_details[p_code][key_rashi]
        rashi_contents[p_idx].append(f"<span class='planet-tag'>{p_code}</span>")

    grid_cells = {}
    for r_idx, (row, col) in SOUTH_GRID_POSITIONS.items():
        rashi_name = RASHIS[r_idx]
        tags = " ".join(rashi_contents[r_idx])
        cell_html = f"""<div class="chart-box"><div class="rashi-label">{rashi_name}</div><div>{tags}</div></div>"""
        grid_cells[(row, col)] = cell_html

    html_out = '<div class="south-chart-container">'
    for r in range(4):
        for c in range(4):
            if (r == 1 or r == 2) and (c == 1 or c == 2):
                if r == 1 and c == 1:
                    html_out += f'<div class="chart-box-center"><div>SOUTH INDIAN<br/><b>{chart_type} CHART</b></div></div>'
            else:
                html_out += grid_cells[(r, c)]
    html_out += '</div>'
    return html_out

# ==============================================================================
# 8. STREAMLIT FRONTEND DASHBOARD
# ==============================================================================
st.markdown('<div class="main-header">🔮 Vedic Astrology Expert System & Dasa Engine</div>', unsafe_allow_html=True)
st.markdown('<div class="sub-header">South Indian Rasi & Navamsa Charts • Dasa-Bhukti Timing • Graha Descriptions • Multilingual Predictions</div>', unsafe_allow_html=True)

# --- SIDEBAR INPUTS ---
st.sidebar.header("📥 Birth Details Input")
birth_date = st.sidebar.date_input("Birth Date", datetime.date(1996, 5, 24))
time_input_str = st.sidebar.text_input("Birth Time (HH:MM in 24-hr format)", "10:30")

try:
    hrs, mins = map(int, time_input_str.split(":"))
    birth_time = datetime.time(hrs, mins)
except Exception:
    st.sidebar.error("Invalid time format. Please enter HH:MM (e.g. 14:30).")
    birth_time = datetime.time(12, 0)

st.sidebar.subheader("📍 Location Selector")
location_type = st.sidebar.radio("Location Mode", ["Popular Cities Database", "Custom Location Search"])

if location_type == "Popular Cities Database":
    selected_city = st.sidebar.selectbox("Select City", list(POPULAR_CITIES.keys()), index=0)
    lat, lon, tz_str = resolve_location(selected_city)
else:
    custom_city = st.sidebar.text_input("Enter City Name (e.g. Colombo, Sri Lanka)", "Colombo, Sri Lanka")
    lat, lon, tz_str = resolve_location(custom_city)

st.sidebar.info(f"**Coordinates:** {lat:.4f}° N, {lon:.4f}° E\n\n**Timezone:** {tz_str}")

local_tz = pytz.timezone(tz_str)
naive_dt = datetime.datetime.combine(birth_date, birth_time)
local_dt = local_tz.localize(naive_dt)
utc_dt = local_dt.astimezone(pytz.utc)

st.sidebar.caption(f"**UTC Calculated Time:** {utc_dt.strftime('%Y-%m-%d %H:%M:%S UTC')}")

# --- CALCULATIONS ---
chart = compute_positions(utc_dt, lat, lon)
moon_deg = chart["MO"]["deg"]
nak_name, birth_lord, remaining_years, dasa_tree = calculate_full_dasa_tree(naive_dt, moon_deg)

# --- TABS INTERFACE ---
tab1, tab2, tab3, tab4, tab5 = st.tabs([
    "📊 Rasi & Navamsa Charts",
    "⚡ Dasa-Bhukti Timeline",
    "🤖 Expert System Life Query",
    "📜 Graha & Yuti Descriptions",
    "📤 Data Lookups & Export"
])

# ------------------------------------------------------------------------------
# TAB 1: CHARTS VIEW
# ------------------------------------------------------------------------------
with tab1:
    col1, col2 = st.columns(2)
    with col1:
        st.subheader("Rasi Chart (D1 - Main Chart)")
        st.markdown(render_south_indian_chart(chart, "D1"), unsafe_allow_html=True)
    with col2:
        st.subheader("Navamsa Chart (D9 - Micro Chart)")
        st.markdown(render_south_indian_chart(chart, "D9"), unsafe_allow_html=True)

    st.divider()
    c1, c2, c3, c4 = st.columns(4)
    c1.metric("Ascendant (Lagna)", chart["LA"]["rashi"])
    c2.metric("Moon Sign (Rasi)", chart["MO"]["rashi"])
    c3.metric("Janma Nakshatra", f"{nak_name} (Pada {chart['MO']['pada']})")
    c4.metric("First Dasa Balance", f"{remaining_years:.2f} Yrs ({birth_lord})")

# ------------------------------------------------------------------------------
# TAB 2: DASA TIMELINE
# ------------------------------------------------------------------------------
with tab2:
    st.subheader("🌳 Vimshottari Dasa-Bhukti-Paryanthar Timeline")

    analysis_date = st.date_input("Target Analysis Date for Timing Queries", datetime.date.today())
    target_dt = datetime.datetime.combine(analysis_date, datetime.time.min)

    active_m, active_b, active_p = find_active_period(dasa_tree, target_dt)

    if active_m and active_b and active_p:
        st.success(f"**Active Period on {analysis_date.strftime('%d-%b-%Y')}:**")
        ac1, ac2, ac3 = st.columns(3)
        ac1.info(f"**Mahadasa Lord:** {active_m['mahadasa_lord']}\n\n`{active_m['start_date'].strftime('%Y-%m-%d')}` to `{active_m['end_date'].strftime('%Y-%m-%d')}`")
        ac2.success(f"**Bhukti Lord:** {active_b['bhukti_lord']}\n\n`{active_b['start_date'].strftime('%Y-%m-%d')}` to `{active_b['end_date'].strftime('%Y-%m-%d')}`")
        ac3.warning(f"**Paryanthar Lord:** {active_p['paryanthar_lord']}\n\n`{active_p['start_date'].strftime('%Y-%m-%d')}` to `{active_p['end_date'].strftime('%Y-%m-%d')}`")

    st.divider()
    st.subheader("Full 120-Year Dasa Hierarchy")
    for m in dasa_tree:
        m_label = f"Mahadasa: {m['mahadasa_lord']} ({m['start_date'].strftime('%b %Y')} - {m['end_date'].strftime('%b %Y')})"
        with st.expander(m_label):
            for b in m["bhuktis"]:
                b_label = f"--- Bhukti: {b['bhukti_lord']} ({b['start_date'].strftime('%d-%b-%Y')} to {b['end_date'].strftime('%d-%b-%Y')})"
                with st.expander(b_label):
                    p_data = [{
                        "Paryanthar Lord": p["paryanthar_lord"],
                        "Start Date": p["start_date"].strftime("%d-%b-%Y"),
                        "End Date": p["end_date"].strftime("%d-%b-%Y")
                    } for p in b["paryanthars"]]
                    st.dataframe(pd.DataFrame(p_data), use_container_width=True)

# ------------------------------------------------------------------------------
# TAB 3: EXPERT SYSTEM & TIMING QUERY
# ------------------------------------------------------------------------------
with tab3:
    st.subheader("🤖 Ask Precise Timing & Life Questions")

    query_domain = st.selectbox(
        "Select Domain or Timing Query:",
        [
            "When will I get promoted / career growth?",
            "When will I get married / find a partner?",
            "Career & Profession",
            "Love, Marriage & Relationships",
            "Wealth & Finances",
            "Education & Intellect",
            "Health & Vitality"
        ]
    )

    user_custom_q = st.text_input("Specific Custom Details / Questions:", "")

    if "prediction_lang" not in st.session_state:
        st.session_state.prediction_lang = "en"

    col_l1, col_l2, col_l3 = st.columns(3)
    if col_l1.button("Read in English 🇬🇧"):
        st.session_state.prediction_lang = "en"
    if col_l2.button("Read in Sinhala 🇱🇰 (සිංහල)"):
        st.session_state.prediction_lang = "si"
    if col_l3.button("Read in Tamil 🇱🇰 (தமிழ்)"):
        st.session_state.prediction_lang = "ta"

    if st.button("Generate Precision Prediction"):
        with st.spinner("Analyzing planetary lords, Dasa timing, and yutis..."):
            ans_bullets = []

            # Retrieve Lagna index
            lagna_rashi_idx = chart['LA']['rashi_idx']

            # Active Dasa timing
            curr_dt = datetime.datetime.now()
            m_act, b_act, p_act = find_active_period(dasa_tree, curr_dt)

            ans_bullets.append(f"### 🌟 Analysis for: {query_domain}")

            # SAFE HOUSE LORD LOOKUPS:
            if "promoted" in query_domain.lower() or "career" in query_domain.lower():
                code_10, name_10 = get_house_lord(lagna_rashi_idx, 10)
                ans_bullets.append(f"**1. Career Timing & House Activation (10th/6th/11th Bhavas):**\nYour 10th House (Karma Bhava) lord is **{name_10} ({code_10})**.")
                if m_act:
                    ans_bullets.append(f"**2. Active Timing Trigger:** Currently running **{m_act['mahadasa_lord']} Mahadasa** and **{b_act['bhukti_lord']} Bhukti**. Favorable promotion windows activate when Dasa lords align with **{name_10}** or 11th house lords.")
            elif "married" in query_domain.lower() or "partner" in query_domain.lower():
                code_7, name_7 = get_house_lord(lagna_rashi_idx, 7)
                ans_bullets.append(f"**1. Relationship Activation (7th/2nd/11th Bhavas):**\nYour 7th House (Kalatra Bhava) lord is **{name_7} ({code_7})**.")
                if m_act:
                    ans_bullets.append(f"**2. Marriage Window:** Active sub-period under **{b_act['bhukti_lord']} Bhukti** activates relationship triggers associated with **{name_7}**.")
            else:
                ans_bullets.append(f"**1. General Domain Strength:** Positioned under **{chart['LA']['rashi']} Lagna** with **{chart['MO']['rashi']} Moon Sign**.")

            # Identify Yutis
            detected_yutis = []
            rashi_groups = {}
            for p_code in ["SU", "MO", "MA", "ME", "JU", "VE", "SA", "RA", "KE"]:
                r_idx = chart[p_code]["rashi_idx"]
                rashi_groups.setdefault(r_idx, []).append(p_code)

            for r_idx, planets in rashi_groups.items():
                if len(planets) >= 2:
                    for p1, p2, name, desc, houses in YUTI_DESCRIPTIONS:
                        if p1 in planets and p2 in planets:
                            detected_yutis.append(f"• **{name} ({p1} + {p2} in {RASHIS[r_idx]}):** {desc}")

            if detected_yutis:
                ans_bullets.append("**3. Active Yuti / Combination Impacts in your Chart:**\n" + "\n".join(detected_yutis))

            full_prediction_text = "\n\n".join(ans_bullets)
            translated_text = translate_prediction(full_prediction_text, st.session_state.prediction_lang)

            st.markdown('<div class="prediction-card">', unsafe_allow_html=True)
            st.markdown(translated_text)
            st.markdown('</div>', unsafe_allow_html=True)

# ------------------------------------------------------------------------------
# TAB 4: GRAHA & YUTI DESCRIPTIONS
# ------------------------------------------------------------------------------
with tab4:
    st.subheader("📜 Individual Graha Descriptions (Bhava 1 to 12)")
    selected_g = st.selectbox("Select Planet / Graha:", list(GRAHA_DESCRIPTIONS.keys()))

    for desc in GRAHA_DESCRIPTIONS[selected_g]:
        st.write(f"• {desc}")

    st.divider()
    st.subheader("🔗 Planetary Combination (Yuti) Matrix")
    df_yuti = pd.DataFrame(YUTI_DESCRIPTIONS, columns=["Planet 1", "Planet 2", "Yoga Name", "Description", "Key Houses"])
    st.dataframe(df_yuti, use_container_width=True)

# ------------------------------------------------------------------------------
# TAB 5: DATA LOOKUPS & EXPORT
# ------------------------------------------------------------------------------
with tab5:
    st.subheader("📊 Detailed Planetary Placements & Dataset Matches")

    data_rows = []
    for p_code in ["SU", "MO", "MA", "ME", "JU", "VE", "SA", "RA", "KE"]:
        p_info = chart[p_code]
        phalaphala = generate_phalaphala(
            p_code, GRAHA_KEYS[p_code], p_info["rashi"],
            p_info["nakshatra"], p_info["pada"], p_info["navamsha_rashi"], p_info["nakshatra_lord"]
        )
        data_rows.append({
            "Code": p_code,
            "Planet": GRAHA_KEYS[p_code],
            "House": f"House {p_info['house']}",
            "Rasi (D1)": p_info["rashi"],
            "Nakshatra": p_info["nakshatra"],
            "Pada": p_info["pada"],
            "Navamsha (D9)": p_info["navamsha_rashi"],
            "Phalaphala Rule Output": phalaphala
        })

    st.dataframe(pd.DataFrame(data_rows), use_container_width=True)
````
